```python
n=np.arange(9)
d=2
r=2
a0=1
g0=1

for i in n:
    plt.plot(i,a0+d*(i),'rs',)
    plt.plot(i,g0*r**i,'bs',alpha=0.5)
plt.legend(labels=['Arithmetic Sequence',
                   'Geometric Sequence'])
nth=6
artseq=a0+d*(nth-1)
geodeq=g0*d**(nth-1)

plt.show()
print(f"the 6th arithmetic sequence is :{artseq} vs \nthe 6th geometric sequence is :{geodeq}\n{artseq==geodeq}")
```

The property of addition and multiplication which states
<hr>
&emsp; that a difference in the order in which numbers are added  <br>
&emsp; &emsp;  or multiplied will not change the result of the operation.  <br>
&emsp; &emsp; &emsp;     For example, 2 + 3 gives the same sum as 3 + 2,  <br>
&emsp; &emsp; &emsp; &emsp;       and 2 × 3 gives the same product as 3 × 2.  <br>

#### distributive property  

```python
3*(3+4)==3*3+3*4
```

```python
# The property which states that
#  for certain mathematical operations, 
#      applying an operation (such as multiplication) to 
#         a set of quantities combined by another operation 
#            (such as addition) yields the same result as 
#               applying the first operation to each quantity individually
#  and then 
#    combining those results by the second operation 
#      (in this case addition).
#      Thus 2 × (3 + 4) is equivalent to (2 × 3) + (2 × 4), 
#        meaning that multiplication is distributive relative to addition.
```

```python
x,y=symbols('x y')
dc=dict(x=[ 4*x-6 ,  x**2-9 ],
        y=[ sympy.sin(y) ])

for keyi in dc:
    print('Solutions for equations involving ' + keyi + ':')
    
    for equ in dc[keyi]:
        
        leftpart=sympy.latex(equ) + '=0' 
        
        midpart= ' \\quad \\Rightarrow \\quad ' + keyi + '='
        
        rightpart=sympy.latex(sympy.solve(equ))
        
        display(Math(leftpart+midpart+rightpart))
    
```

##  Factor an expression

### prime factorization 

```python
# whats a prime number 
#    its a number can only be divisable by it self and i 
#      eg : 2,3,5,7,
```

```python
# 8=2x2x2
# 25=5x5
# 48=2x2x2x2x3
```

```python
number =188
sympy.factorint(number)
print(sympy.factorint(number))
type(sympy.factorint(number))
sympy.factorint(number).keys()

# how to interpretate the output - 
#  since the solution is wrapped within a dictionary 
#    the first key is the firt portion of the prime factor 2 and the value means how many times 
#      the first key multiplies by itself  
#         thus how many keys indicates how many prime factors are there , 
#           and each value in the key_value pair is the times the key multiplies by itself 

#  therefore, 
#  {2:4 ,3:1}
#  can be interpretated as 2x2x2x2 x 3
#                          --2**4----3**1----
```

```python
dct=sympy.factorint(number)
print(f"keys : {dct.keys()}")
print(f"keys : {dct.items()}")
print(f"keys : {dct.values()}")
```

```python
for i in range(2,51):
    pf=list(sympy.factorint(i).keys())
    if pf[0]!=i:
        print(f"{i} is composite number with prime factors:{list(sympy.factorint(i).keys())}")
    else:
        print(f"{i} is a prime number")
# prime number can only be divisable by itself and 1 , thus if the number's factorint keys [0]
# is not itselft can be classified as composite number
```

```python
for i in range(2,51):
    pf=(sympy.factorint(i))
    ki=list(pf.keys()) 
    # print(pf[ki[0]])
    # print(pf)
    if len(pf)==1 and pf[ki[0]]==1:
        # one key_value pair 's len is 1 
                      # key[0]==1 means the prime number can only be divisable by 1 (
                      # if the second condition isnt 1 means that the number can be divisable by itself n times 
                      # reverse the process, although the value_key pair is only 1, but the combination can 
                      # produce a non prime number as the key might be any number than 1. 
                      # therefore, if the key value is not 1 , then the value will be mutiplied by itself by n time 
                      # the first condition confines the chance of being divisable by itself 
        print(f"{i} is a prime number")
    else:
        print(f"{i} is composite number with prime factors:{list(sympy.factorint(i).keys())}")
```

### Reduce fractions to lowest terms

```python
numerator = 12
denominator = 24
factor = 1

allfactors=[]
# Find greatest common factor 
for test_factor in range(1,denominator+1): 
    if numerator%test_factor==0 and denominator%test_factor==0:
        allfactors.append(test_factor)
        factor = test_factor  # updated to the last common factor in sequence 
print(f"The greatest common factor is : {factor}")
print(f"All the common factors are  : {allfactors}")
# Divide out greatest common factor 
n = int(numerator/factor) 
d = int(denominator/factor) 

print("original: ", numerator, "/", denominator) 
print("reduced: ", n, "/", d) 
```

```python
n = 48 # input (factor out a sqrt )
upper_limit = math.floor(math.sqrt(n)) + 1 # why to find out the maximum factor 
for maybe_factor in range(1, upper_limit): 
    if n % (maybe_factor**2) == 0: 
        max_factor = maybe_factor**2 
other_factor = int(n/max_factor) # integer inside the sqrt 
square_root = int(math.sqrt(max_factor)) # why - for loop **2
output = square_root*sympy.sqrt(other_factor) 
output #can't use print function if the formula is to be preserved 
```

##### multipilication conjugate

$$z = a \quad bi $$
$$w = c \quad di $$
complex conjugation -using distributive property
$$z \cdot w = (a \quad bi) \cdot (c \quad di) $$
$$ \quad \quad \quad \quad \quad=ac+adi+cbi+bdi^2$$
because 
$$ i^2 =-1 \quad i= \sqrt{-1}$$
$$ \quad   \quad \quad =ac+adi+cbi+bd$$

$$-\text{complex number}|    \quad\quad   |\text{complex conjugate}$$
$$a\quad +bi|\quad\quad|a\quad -bi$$
$$a\quad -bi|\quad\quad|a\quad +bi$$
only the imaginary part's sign is flipped , when converting complex number to complex conjugate

complex conjugation -using distributive property
$$z \cdot z^*= (a \quad bi) \cdot (c \quad-bi) $$
$$~ \quad \quad \quad  = a^2-abi+abi-(bi)^2$$
$$~=a^2-(bi)^2\quad\quad$$
$$=a^2+b^2\quad\quad\quad$$

###  Quadratic Functions - roots, vertext, graph

#### def

what's a quadratic equation $$ $$
its a polynomial thats set to equal to zero 

$$ax^2+bx+c=0 $$ 
Its simplily a degree two polynomial equation 
$$ $$
how do we find the solution to this quadratic equation 
$$x=\frac{-b\pm \sqrt{b^2-4ac}}{2a} $$

vertex formula
$$h=-\frac{b}{2a} \quad k=f(h)$$
$$f(x)=a(x-h)^2+k $$

#### workflow

```python
# vertex = -b/2a 
# cross x axix -then y =0 solving the (0,-x1)(0,x2)  
#  a pair of possible answers reason- because x is sqrt which yields 
#    both a negative and a positive value 
# quadratic formula 
#   -b/2a      +-  sqrt(b^2+4ac)/2a 
#    vertex        pq - middle 
#  middle point 
```

#### simple solution

```python
q_a = 2
q_b = 7
q_c = 5

quaP=(-q_b+np.sqrt(q_b**2-(4*q_a*q_c)))/(2*q_a)
quaN=(-q_b-np.sqrt(q_b**2-(4*q_a*q_c)))/(2*q_a)

print(quaP,quaN)
```

#### solve & plot

```python
import matplotlib.pyplot as plt
import numpy as np
import math

# Define the three coefficients here
a = 1
b = 5
c = 6

print("y = ", a, "x**2 + ", b, "x + ", c) 

# Vertex
vx = -b/(2*a) 
vy = a*(vx**2) + b*vx + c 
print("Vertex: (",vx,",",vy,")")

# Roots
d = b**2 - 4*a*c # d -(determinent)
if d>=0: 
    root_1 = (-b + math.sqrt(d))/(2*a) 
    root_2 = (-b - math.sqrt(d))/(2*a) 
    print("Roots: x = ", root_1, " and x = ", root_2)
else:
    print("No real roots")

xmin = -10
xmax = 10
ymin = -10
ymax = 10
points = 10*(xmax-xmin)
x = np.linspace(xmin,xmax,points) 

fig, ax = plt.subplots()
plt.axis([xmin,xmax,ymin,ymax]) # window size
plt.plot([xmin,xmax],[0,0],'b') # blue x axis
plt.plot([0,0],[ymin,ymax], 'b') # blue y axis

# Graph the parabola
y1 = a*x**2 + b*x + c 
plt.plot(x, y1) 

# Plot the vertex
plt.plot([vx],[vy], 'ro') 
#Here, 'r' stands for red color, and 'o' stands for circle markers.

# Plot the roots
plt.plot([root_1, root_2],[0,0], 'go') 

plt.show() 
```

#### plot with sliders

```python
%matplotlib inline
from ipywidgets import interactive
import matplotlib.pyplot as plt
import numpy as np
import math

# All graphing happens in this function
def f(a,b,c):
    xmin = -10
    xmax = 10
    ymin = -10
    ymax = 10
    points = 10*(xmax-xmin)
    x = np.linspace(xmin,xmax,points)

    plt.axis([xmin,xmax,ymin,ymax]) # window size
    plt.plot([xmin,xmax],[0,0],'b') # x axis
    plt.plot([0,0],[ymin,ymax], 'b') # y axis

    # Parabola
    y1 = a*x**2 + b*x + c 
    plt.plot(x, y1) 

    # Vertex
    vx = -b/(2*a) 
    vy = a*(vx**2) + b*vx + c 
    plt.plot([vx],[vy], 'ro') 

    # Roots
    d = b**2 - 4*a*c 
    if d>=0: 
        root_1 = (-b + math.sqrt(d))/(2*a) 
        root_2 = (-b - math.sqrt(d))/(2*a) 
        plt.plot([root_1, root_2],[0,0], 'go') 
    
    # Set the equation as the title
    sa = str(a)
    sb = str(b)
    sc = str(c)
    h1 = "y = ", sa, "x**2 + ", sb, "x + ", sc  # concatenation 
    h2 = ""
    
    for w in h1:
        h2 = h2 + w

    plt.title(h2,fontsize=22,fontweight='bold',color='indigo')
    plt.show()

# Interactive creates sliders for a, b, and c values
interactive_plot = interactive(f, a=(1, 9), b=(-9,9), c=(-9,9))
#                           interactive (function , parameter (raneg))
interactive_plot


```

```python
def quadraeq(a,b,c):
    qs=np.zeros(2)
    qs[0]=(-b+np.sqrt(b**2-(4*a*c)))/(2*a)
    qs[1]=(-b-np.sqrt(b**2-(4*a*c)))/(2*a)
    return list(qs)
    
a=1
b=range(-5,6)
c=range(-2,11)

# qsexer=(-b_e + np.sqrt(b_e**2-4*a_e*c_e)) /(2*a_e)
M=np.zeros((len(b),len(c)))
M.shape

for bi in range(0,len(b)):
    for ci in range(0,len(c)):
      M[bi,ci] = quadraeq ( a, b[bi]  , c[ci])[0]
        
for bi in range(0,len(b)):
    for ci in range(0,len(c)):
        if M[bi,ci]>0.0000000000001 :
            plt.text(ci,bi,s=int(M[bi,ci]))
        if M[bi,ci]<-0.000000000001:
            plt.text(ci,bi,s=int(M[bi,ci]))
plt.imshow(M)
plt.show()
```

#### line slope 

```python
display(Math('y=mx+b'))
```

$$y=mx+b$$
$$m=slope=\frac{rise}{run}=\frac{\Delta{x}}{\Delta{y}}$$
$$b=y-intercept$$
solving b : set y=0
$$mx+b=0 -------x=-\frac{b}{m} $$
solving m 
$$m=\frac{b-y}{x} $$

```python
h_m_c=2
k_m_c=-3
r_m_c=3

# radius + origin's coordinates, 
#  np.max() a : array_like Input data. therefore those two values that specify the 
#   origin's x and y values are passed in the form of a tuple -which is a collection of data points 
axlim = r_m_c + np.max((abs(h_m_c),abs(k_m_c)))
x_m_c = np.linspace(-axlim,axlim,100)
# use the limit to generate the range of linspace upper and lower bounds , which 
# is more efficient in defining the canvas size which is the figsize itself,
# yet the circle may not be placed within the center 

X_M_C,Y_M_C = np.meshgrid(x_m_c,x_m_c)

# circle formula rearranged by moving the r^2 to the right side without explicitly assign a 0 to the euqation 
# pythagrean's theorem 
FXY_M_C=(X_M_C - h_m_c)**2 + (Y_M_C - k_m_c)**2 - r_m_c**2

plt.contour(FXY_M_C)
```

#### parabola

##### def 

```python
# parabola is a set of all points with equal distance from
# a focus point and a line(directrix)
```

```python
# baseball being thrown 's  trajectory 
# satelite dishes (receiver/transmitter is at the center of the parabola )
# magnifier
# 
# parabola equation derivation 
```

basic parabola
$$ y=x^2$$
general form
$$ y-k=a(x-h)^2 \quad \quad \text{opens +a \quad up \quad-a \quad \quad down}$$
$$ x-h=a(y-k)^2 \quad \quad \text{opens +a \quad right -a \quad left}$$
$$ \text{ !!!!!!  the vertex is always halfway between a focus point and directrix line } $$
$$ \text{ !!!!!!  the focus  is always inside the parabola which means it s always positioned above the vertex} $$
$$ a=\frac{1}{4c} $$
$$ \text{Vertex}= (h,k) $$
$$ \text{Focus}= (h,k+c / p / \frac{a}{4}) $$
$$ \text{Directrix}: y= k-c $$
$$ \text{Axis of symmetry}: x=h $$

k shift in y h shift in x 

a/4p is how wide the parabola 

the larger the a value or /4p , the narrower the parabola opens

if a/p is negative opens down,left , positive up,right

elipse has two focus points , 

hyperbola has two ---------

parabola has only one---------- 

##### expression 

$$4p(y-k)=(x-h)^2 $$
shift k and h 
$$y=\frac{1}{4p}(x-h)^2 +k$$

##### types -vertical horizontal 

```python
para_a=1
para_h=1
para_k=-2
para_n=15
para_x=np.linspace(para_h-5,para_h+5,para_n)
para_y=para_a*(para_x-para_h)**2 + para_k

plt.plot(para_x,-para_y,'--',color='indigo',alpha=0.3)
plt.plot(para_x,para_y,'-.',color='blue',alpha=0.3)

plt.plot(para_y,para_x,'-.',color='cyan',alpha=0.3)
plt.plot(-para_y,para_x,'--',color='red',alpha=0.3)

plt.text(-4,20,s='(x-h)^2=4p(y-k)')
plt.text(0,15,s='+ p',color='blue')
plt.text(0,-15,s='- p',color='indigo')
plt.text(-20,-10,s='(y-k)^2=4p(x-h)')
plt.text(20,0,s='- p',color='cyan')
plt.text(-20,0,s='+ p',color='red')

plt.show()
```

#### elipses

#### equation derivation using circle's equation

circle equation (origin's coordinates are (0,0)) no shift in both x and y direction 
$$ x^2+y^2=r^2 $$
substitute r with 3 constant 
$$x^2+y^2=3^2 $$
divide both sides of the equation by a factor of 3^2
$$\frac{x^2+y^2}{3^2}=\frac{3^2}{3^2} $$
simplify both sides 
$$\frac{x^2}{3^2}+\frac{y^2}{3^2}=1 $$

at this stage the euqation of a circle is simiar with
the one of
an elipse with the only exception of identical denominator

$$\frac{x^2}{a^2}+\frac{y^2}{b^2}=1 $$

a > b x direction stretch $$ ~ $$
a < b y direction stretch

##### mannual graphing workflow

###### intercept

$$\frac{x^2}{3^2}-\frac{y^2}{4^2}=1 $$
set y =0
$$ \frac{x^2}{3^2}=1 $$
$$ x^2=3^2 $$
$$x_1=+3 \quad x_2=-3 $$
y has no intercept
$$ y^2=-16 $$

###### solving y and plugging x

$$\frac{x^2}{3^2}-\frac{y^2}{4^2}=1 $$
$$-\frac{y^2}{4^2}=1 -\frac{x^2}{3^2}$$
$$-\frac{y^2}{4^2}= -\frac{x^2}{3^2}+1$$
$$ y^2=-16(-\frac{x^2}{9}+1) $$
$$ y=\sqrt{16\cdot (\frac{x^2}{9}-1)} $$
import step in symplifing the equation - factoring a fraction from the sqrt
$$y=\sqrt{\frac{16}{9}\cdot(x^2-9)} $$
$$y=\frac{4}{3}\cdot\sqrt{x^2-9} $$

```python
fig,ax=plt.subplots(figsize=(8,6))
### hyperbola ----------------------------
for i in range(-20,20):
    plt.plot(i,(4/3)*np.sqrt(i**2-9),'o')
    plt.plot(i,-(4/3)*np.sqrt(i**2-9),'o')

### asymptote ----------------------------
plt.plot(np.arange(-20,20),
         (4/3)*np.arange(-20,20),'--',
         color='black')
plt.plot(np.arange(-20,20),
         -(4/3)*np.arange(-20,20),'--',
         color='black')

# x intercept ------------
plt.plot(-3,0,'b*',ms=40,alpha=0.3)
plt.plot(3,0,'b*',ms=40,alpha=0.3)

ax.set_ylim(-15,15)
ax.set_xlim(-10,10)
plt.show()    
```

$$y=\frac{4}{3} \cdot \sqrt{x^2-9} $$
$$y=\frac{4}{3} \cdot \sqrt{x^2\cdot( 1- \frac{9}{x^2})} $$
As x approaches infinity 
$$ y \approx \pm \frac{4}{3} \cdot x$$

#### elipse vs hyperbola vs circle

conic section formula summary
$$\text{circle    formula is :}  (y-k)^2+(x-h)-r^2=0\quad$$
$$ \text{parabola formula is :}  (y-k)^2=4p(x-h) \quad  x^2 \quad\text{updown}\quad y^2 \quad\text{rightleft}\quad$$
$$\text{elipse    formula is :}  \frac{(y-k)^2}{b^2}+\frac{(x-h)^2}{a^2}-1=0$$
$$\quad \quad \text{hyperbola formula is :}  \frac{(y-k)^2}{b^2}-\frac{(x-h)^2}{a^2}-1=0$$

$$\text{circle    formula is :}  y^2 + x-r^2=0\quad$$
$$ \text{parabola formula is :}  y^2=x \quad  x^2 \quad\text{updown}\quad y^2 \quad\text{rightleft}\quad$$
$$\text{elipse    formula is :}  \frac{y^2}{b^2}+\frac{x^2}{a^2}-1=0$$
$$\quad \quad \text{hyperbola formula is :}  \frac{y^2}{b^2}-\frac{x^2}{a^2}-1=0$$

##  Solve for a variable

```python
from sympy import *
import math

# Identify all variables
var('a b c d x') 

# Left and right sides of the equal sign
left = 0
right = a*x**2 + b*x + c

# Variable to solve for
variable = x

# Sympy equation left = right
eq1 = Eq(left,right) 

# Sympy solve for that variable
sol = solve(eq1,variable) 

# Show factored results
for s in sol:
    print(variable, " = ", s)
```

### work flow 

### find equation

## Graph lines and zoom in or out

```python
%matplotlib inline
from ipywidgets import interactive
import matplotlib.pyplot as plt
import numpy as np

# Define the graphing function
def f(m,b,zoom):
    xmin = -zoom
    xmax = zoom
    ymin = -zoom
    ymax = zoom
    points = 10*xmax-xmin
    x = np.linspace(xmin, xmax, points)

    fig, ax = plt.subplots()
    plt.axis([xmin,xmax,ymin,ymax]) # window size
    plt.plot([xmin,xmax],[0,0],'black') # black x axis
    plt.plot([0,0],[ymin,ymax], 'black') # black y axis
    
    # Change y1 to be any function
    y1 = m*x+b
    plt.plot(x, y1)
    # Plot another line here if you want
    # y2 = 
    # plt.plot(x, y2)    
    
    ax.set_xlabel("x values")
    ax.set_ylabel("y values")
    ax.set_title("Some Graph")
    ticks = int(round((xmax-xmin)/20))
    ax.set_xticks(np.arange(xmin, xmax, ticks))
    ax.set_yticks(np.arange(ymin, ymax, ticks))
    ax.grid(True)
    plt.show()

# Set up the slider
interactive_plot = interactive(f, zoom=(1,100),b=(-40,40),m=(-9,9))
interactive_plot
```

##  Solve & graph a system equations

```python
from sympy import *
from sympy.plotting import plot
from sympy import sqrt

var('x y') 

# First equation set equal to zero, ready to solve
first = 2*x+y-1


# Second equation set equal to zero, ready to solve
second = x-2*y+7

# 
# Solve - can be linear or nonlinear equations
solution = linsolve([first, second], (x, y)) 
print(solution,
      f"args {solution.args}",
      str(solution).split(","),
      str(solution).split(",")[0],
      str(solution).split(",")[1],
      sep="\n")


for a in range(len(solution.args)):
    x_solution = solution.args[a][0] 
    y_solution = solution.args[a][1] 
    print("Solution = (", x_solution, ",", y_solution, ")")

# Sympy syntax for equation equal to zero, ready to factor
y_first = Eq(first,0) 

# Sympy solve for y
y1 = solve(y_first,y) 

# Same two steps for second equation
y_second = Eq(second,0) 
y2 = solve(y_second,y)

# Show factored results
print("y = ", y1[0])
print("y = ", y2[0])

# Plot solution
x = symbols('x')
xmin = -10
xmax = 10
plot(y1[0], y2[0], (x,xmin,xmax))  

```

Nicer looking output

```python
from sympy import *

x,y = symbols('x y')

first = 2*x + y - 1 
second = x - 2*y + 7 


# parse finite set answer as coordinate pair
solution = linsolve([first, second], (x, y)) 
x_solution = solution.args[0][0] 
y_solution = solution.args[0][1] 

print(f"first {solution.args[0][0]}",
      f"second {solution.args[0][1]}",
      f"arg-first {solution.args[0]}",
      f"arg {solution.args}",
      f"sol {solution}",
      sep='\n') 

#In Python, a finite set refers to a collection of distinct, 
#unique elements that has a specific, countable number of members

# Print a coordinate pair
print("(xsolution is :",x_solution,"\n", "ysolution is :",   y_solution, ")") 
```

Solve and graph. Factor to change from "`0 = `" to "`y = `" preparing to graph 

```python
from sympy import *
from sympy.plotting import plot
from sympy import sqrt

var('x y') 
# First equation set equal to zero, ready to solve
first = -x**2 - y + 10
#Second equation set equal to zero, ready to solve
second = 2*x**2 - 2*y - 4

#--------------------solving x and y ----------------------
# Solve - can be linear or nonlinear equations
solution = nonlinsolve([first, second], (x, y)) 
for a in range(len(solution.args)):
    x_solution = solution.args[a][0] 
    y_solution = solution.args[a][1] 
    print("Solution = (", x_solution, ",", y_solution, ")")
    
#--------rearrange formula by moving y the left side  ------
# Sympy syntax for equation equal to zero, ready to factor
# Sympy solve for y
y_first = Eq(first,0) 
y1 = solve(y_first,y) 
y_second = Eq(second,0) 
y2 = solve(y_second,y)   # [{y: x**2 - 2}] if y is not specified 

# Show factored results
print("y = ", y1[0])  # [0]removing the square brackets 
print("y = ", y2[0])
# similar to f(X)= 

# Plot solution
x = symbols('x')
xmin = -10
xmax = 10
plot(y1[0], y2[0], (x,xmin,xmax))  
plt.show()
```

```python
# sliders and zoom
# y = np.floor(x)

%matplotlib inline
from ipywidgets import interactive
import matplotlib.pyplot as plt
import numpy as np

# Define the graphing function
def f(a,b,c,zoom):
    xmin = -zoom
    xmax = zoom
    ymin = -zoom
    ymax = zoom
    points = 10*xmax-xmin
    x = np.linspace(xmin, xmax, points)

    plt.axis([xmin,xmax,ymin,ymax]) # window size
    plt.plot([xmin,xmax],[0,0],'black') # black x axis
    plt.plot([0,0],[ymin,ymax], 'black') # black y axis
    
    # Line 1
    y1 = a*np.floor(x-b) + c
    plt.plot(x, y1, 'ro')
    plt.show()

# Set up the sliders
i_plot=interactive(f,a=(-9,9),b=(-9,9),c=(-9,9),zoom=(1,100))
i_plot
```

## Calculate the price elasticity of demand
$ \frac{percent\ change\ in\ demand}{percent\ change\ in\ price} $


The degree of buyers' responsiveness to price changes. Elasticity is measured as the percent change in quantity divided by the percent change in price. A large value (greater than 1) of elasticity indicates sensitivity of demand to price, e.g., luxury goods, where a rise in price causes a decrease in demand. Goods with a small value of elasticity (less than 1) have a demand that is insensitive to price, e.g., food, where a rise in price has little or no effect on the quantity demanded by buyers.

```python
old_demand = 5
new_demand = 6
old_price = 11
new_price = 10

# percent change in price
price_change = (new_price - old_price) / old_price


# percent change in demand
demand_change = (new_demand - old_demand) / old_demand

e_number = demand_change/price_change

# elasticity
print(e_number)

if e_number>1:
    print("Demand is elastic")
elif e_number==1:
    print("Demand is unitary or proportional")
elif e_number<1:
    print("Demand is inelastic")
else:
    print("?")
```

## Calculate Price Elasticity of Supply
$ \frac{percent\ change\ in\ supply}{percent\ change\ in\ price} $

```python
old_price = 4
new_price = 7
old_supply = 10
new_supply = 11


# percent change in price
price_change = (new_price - old_price) / old_price


# percent change in supply
supply_change = (new_supply - old_supply) / old_supply

e_number = supply_change/price_change

# elasticity
print(e_number)

if e_number>1:
    print("Supply is elastic")
elif e_number==1:
    print("Supply is unitary or proportional")
elif e_number<1:
    print("Supply is inelastic")
else:
    print("?")
```

##  Compound interest calculator



When the compunding happens <i>n</i> times per year:<br>
Annuity = P(1 +  $\frac{r}{n})^{nt}$<p>
When the growth is continuous: <br>
Annuity = Pe<sup>rt</sup>

```python
import math

# Change these variables
p = 1
r = .01
t = 1
n = 1

n_times = p*(1+(r/n))**(n*t)
continuous = p*math.e**(r*t)

print("Annuity when compounded ", n, " times per year = ", n_times)
print("Annuity when compounded continuously = ", continuous)
```

## Investment annuity estimator



Calculte the value in a retirement account, given regular monthly contributions and an estimated rate of increase

```python
# Change these variables
p = 1000
r = .08
t = 38
monthly = 350

# The loop will update this variable
annuity = p

# loop for contribution and percent growth
for a in range(12*t):
    annuity = (annuity+monthly)*(1+(r/12))

print("annuity = ", round(annuity, 2))
print("Annual income from interest = ", round(annuity*r, 2))
```

##  Mortgage payment calculator


Mortgage payment formula:<p>
monthly payment = P$\frac{\frac{r}{12}(1 + \frac{r}{12})^{12t}}{(1 + \frac{r}{12})^{12t} - 1}$ 


```python
# Change these values
p = 240000
r = .055
t = 30

# Break down that formula into parts
numerator = (r/12)*(1+(r/12))**(12*t) 
denominator = (1+(r/12))**(12*t) - 1 
payment = round(p*numerator/denominator, 2)

print("monthly payment = ", payment)
```

Payment function without rounding

```python
def payment(p,r,t):
    numerator = (r/12)*(1+(r/12))**(12*t)
    denominator = (1+(r/12))**(12*t) - 1
    payment = p*numerator/denominator
    return payment

print(payment(200000,.06,30))
```

Similar function with rounding to compare payments for slight changes in principle

```python
%matplotlib inline
from ipywidgets import interactive

def pmt(p):
    r = .06
    t = 30
    numerator = (r/12)*(1+(r/12))**(12*t)
    denominator = (1+(r/12))**(12*t) - 1
    pay = round(p*numerator/denominator, 2)
    print("Principle: ", p, " Payment: ", pay)

show_slider = interactive(pmt, p=(0,300000))
show_slider
```

##  mortgage - annual change 

```python
p = 240000.00
r = .055
t = 30


# Break down that formula into parts
numerator = (r/12)*(1+(r/12))**(12*t)
denominator = (1+(r/12))**(12*t) - 1
payment = round(p* numerator/denominator, 2)

print("payment = ", payment)

# loop to show interest amounts
balance = p
print("month \t balance \t interest") 
for a in range(12*t):
    interest = round(balance*r/12, 2)
    if a%24==0 or a==359:
        print(a, "\t", balance, "\t", interest) 
    balance = round(balance+interest-payment,2) 
```

### log laws 

#### mul +

$$log_{10}^{\quad a \cdot b}=log_{10}^{\quad a }+log_{10}^{\quad b } $$
$$log^{ a \cdot b}=log^{ a }+log^{ b } $$

#### div -

$$\log_{10}^{\quad \left(\frac{a}{b}\right)}=\log_{10}^{\quad a }-\log_{10}^{ \quad b } $$
$$\log^{\left(\frac{a}{b}\right)}=\log^{a}-\log^{b}$$

## Draw Lines

Simple example

### triangle plt(path&patch)


```python
import matplotlib.pyplot as plt
from matplotlib.path import Path
import matplotlib.patches as patches

verts = []
steps = []

# add starting point
verts.append((0.,0.))
steps.append(Path.MOVETO)

# add lines one at a time
verts.append((2.,2.))
steps.append(Path.LINETO)

# more
verts.append((2.,-2.))
steps.append(Path.LINETO)

verts.append((0.,0.))
steps.append(Path.CLOSEPOLY)

# graph
fig, ax = plt.subplots()
plt.axis([-5,5,-5,5])
path = Path(verts, steps)
patch = patches.PathPatch(path)
ax.add_patch(patch)
plt.show()
```

More elaborate arrays

### polygon  plt(path&patch)

```python
import matplotlib.pyplot as plt
from matplotlib.path import Path
import matplotlib.patches as patches

# array of points
verts = [(0., 0.), (-1., 1.), (0., 2.), (2., 2.), (3., 1.), (2., 0.), (0., 0.),]

# array of steps
codes = [Path.MOVETO,
         Path.LINETO,
         Path.LINETO,
         Path.LINETO,
         Path.LINETO,
         Path.LINETO,
         Path.CLOSEPOLY, ]

# graph
fig, ax = plt.subplots()
plt.axis([-5,5,-5,5])
path = Path(verts, codes)
patch = patches.PathPatch(path, fill=False,facecolor='blue', lw=2)
ax.add_patch(patch)
plt.show()

```

#Python trig functions
These functions use radians, so you have to use the `math.radians()` function inside the sine, cosine, or tangent functions.

## trigonometry

### radiant vs degree

```python
plt.plot(np.cos(np.linspace(0,2*np.pi)),
         np.sin(np.linspace(0,2*np.pi)))
plt.axis('square')
plt.show()
```

$$ 2 \cdot \pi \quad\text{radians} $$
$$ 360^o \quad \text{degrees} $$
$$ 2 \cdot \pi =360^o $$
conversion from degrees to radians
$$ rad=\frac{360}{2\pi}=\frac{180}{\pi} $$
conversion from radians to degrees
$$ deg=\frac{2\pi}{360}=\frac{\pi}{180} $$

### rad2deg converter

```python
degrees_to_be_converted=90  # input is needed
 
rad_result= np.round(degrees_to_be_converted*np.pi/180,2)

radians_to_be_converted=6   # input is needed

deg_result=np.round((radians_to_be_converted*180)/np.pi,0)

print(f"input radian=({degrees_to_be_converted}) converted to deg_result is ------{rad_result},\ninput degrees=({radians_to_be_converted}) rad_result is------{deg_result}")
```

### pythagrean theorem

$$c^2=a^2+b^2 $$

```python
py_a=3
py_b=4

fig,ax=plt.subplots()
py_c=np.sqrt(py_a**2+py_b**2)


plt.plot([0,py_a],[0,0],'k')
plt.plot([0,0],[0,py_b],'r')
plt.plot([0,py_a],[py_b,0],'g')

plt.plot([0,-py_a],[0,0],'k')
plt.plot([0,0],[0,-py_b],'r')
plt.plot([0,-py_a],[-py_b,0],'g')
rec_py=matplotlib.patches.Rectangle((0,0),0.4,0.4,facecolor='white',edgecolor='black')
ax.add_patch(rec_py)
plt.axis('square')
py_axlim=np.max((py_c,py_b))+2



plt.axis([-py_axlim,py_axlim,
         -py_axlim,py_axlim])
plt.axis('off')


py_xcir=np.linspace(-10,10,100)

for i in py_xcir:
    
   plt.plot(i,np.sqrt((py_b**2)-(i**2)),'*',ms=2)
   plt.plot(i,-np.sqrt((py_b**2)-(i**2)),'v',ms=2)

# text----------------------
plt.text(1,1.1,
         s='Hypotenuse',
         color='green',
         rotation=-55,
         fontsize=10)
plt.text(-0.6,1,
         s='opposite',
         color='red',
         rotation=90,
         fontsize=10)
plt.text(0.5,-0.6,
         s='ajacent',
         color='black',
         rotation=0,
         fontsize=10)
# text----------------------
plt.plot([0,py_a],[-py_b,0],'g')
plt.plot([0,-py_a],[py_b,0],'g')
plt.show()
```

### sin cos tan

##### np.sin(x)

```python
s_c_x=np.linspace(0,2*np.pi,60)

for i in s_c_x:
    plt.plot(i,np.sin(i),'o')
    plt.plot(i,np.cos(i),'*')
plt.legend(['$sin(\\theta)$','cos($\\theta$)'])
plt.show()
```

##### np.cos(x)

```python
s_c_x=np.linspace(0,2*np.pi,60)

for i in s_c_x:
    plt.plot(i,np.sin(i),'o')
    plt.plot(i,np.cos(i),'*')
plt.legend(['$sin(\\theta)$','cos($\\theta$)'])
plt.show()
```

##### np.tan(x)=sin/cos

```python
s_c_x=np.linspace(0,2*np.pi,40)

for i in s_c_x:
    plt.plot(i,np.tan(i),'v')
plt.plot(s_c_x,np.tan(s_c_x),'--',color='cyan')
plt.legend(['$tan(\\theta)$','$tan(\\theta)$'])
plt.show()
```

```python
tri_x0=np.linspace(0,2*np.pi,40)
tri_y1=np.sin(tri_x0+np.cos(tri_x0))
tri_y2=np.cos(tri_x0+np.sin(tri_x0))

for i in range(len(tri_x0)):
    plt.plot(tri_x0[i],tri_y1[i],'o',alpha=0.5)
    plt.plot(tri_x0[i],tri_y2[i],'*',ms=10)
plt.legend(['y1=sin( $\\theta$+cos( $\\theta)$)',
            'y2=cos( $\\theta$+sin( $\\theta)$)'    ],
           bbox_to_anchor=(0.9,1))
plt.savefig('ss.pdf')
plt.show()
```

##### 7pi/6

```python
tri_x22=np.linspace(0,2*np.pi,100)

fig,ax=plt.subplots(figsize=(15,10))
######### full circle #################################
plt.plot(np.cos(tri_x22),
         np.sin(tri_x22))
######### arc #########################################
tri_x23=np.linspace(0,(7*np.pi)/6,40)
plt.plot(np.cos(tri_x23),
         np.sin(tri_x23),color='black',linewidth=5)
########## dots #####################################
plt.plot(np.cos((7*np.pi)/6),
         np.sin((7*np.pi)/6),'ro',ms=10)
plt.plot(np.cos(0),
         np.sin(0),'bo',ms=10)
########## center line #############################
plt.plot([0,0],[-1,1],'k--')
plt.plot([-1,1],[0,0],'k--')
#########  trig ####################################
plt.plot([0,np.cos((7*np.pi)/6)],
         [0,np.sin((7*np.pi)/6)],'--')

plt.plot([0,np.cos((7*np.pi)/6)],
         [0,0],color='green',linewidth=3,label='cos part')

plt.plot([0,0],
         [0,np.sin((7*np.pi)/6)],
         color='red',linewidth=3,label='sin part')

plt.plot([0,np.cos((7*np.pi)/6)],
         [np.sin((7*np.pi)/6),np.sin((7*np.pi)/6)],'y--')
plt.plot([np.cos((7*np.pi)/6),np.cos((7*np.pi)/6)],
         [np.sin((7*np.pi)/6),0],'g--')

######## finishing touches #######################
plt.title('$\\theta=\\frac{7\\pi}{6}$',
          fontsize=15,color='red')
plt.legend()
plt.axis('square')

plt.show()
```

### Euler's formula

$$e^{ik}=cos(k)+sin(k) $$
Euler's Identity
$$e^{i\pi}+1=0 $$

```python
tri_x22=np.linspace(0,2*np.pi,100)

fig,ax=plt.subplots(figsize=(10,8))
######### full circle #################################
plt.plot(np.cos(tri_x22),
         np.sin(tri_x22))
######### arc #########################################
tri_x23=np.linspace(0,(np.pi),40)
plt.plot(np.cos(tri_x23),
         np.sin(tri_x23),color='black',linewidth=5)
# ########## dots #####################################
plt.plot(np.cos((np.pi)),
         np.sin((np.pi)),'ro',ms=20)
plt.plot(np.cos(0),
         np.sin(0),'bo',ms=20)

plt.text(np.cos((np.pi))+0.1,
         np.sin((np.pi))+0.1,
         color='red',
         s='cos($\pi$)=-1',
         fontsize=20)
plt.text(np.cos(0)-0.5,
         np.sin(0)+0.1,
         color='blue',
         s='cos(0)=1',
         fontsize=20)

plt.text(np.cos(np.deg2rad(90))-0.15,
         np.sin(np.deg2rad(90))+0.03,
         color='indigo',
         s='sin(90)=1',
         fontsize=20)
plt.text(np.cos(np.deg2rad(270))-0.2,
         np.sin(np.deg2rad(270))-0.08,
         color='cyan',
         s='sin(270)=-1',
         fontsize=20)
########## center line #############################
plt.plot([0,0],[-1,1],'k--')
plt.plot([-1,1],[0,0],'k--')
plt.title('$e^{ik}$=cos(k)+sin(k)',fontsize=20,color='red')
plt.text(-0.8,0.3,s='$e^{i\\pi}\quad + \quad 1=0  $',fontsize=20,color='indigo')
# plt.legend()
plt.axis('square')
plt.savefig('y.pdf')
plt.show()
```

$$e^{ik}=cos(k)+sin(k) $$

##### spoke plt.polar

##### spiral plt.polar

```python
# create an outwards spiral using 
# phase angles and amplitudes
spiral_n = 100
spiral_a = np.linspace(0,1,spiral_n)
spiral_p = np.linspace(0,4*np.pi,spiral_n)

for i in range(100):
    plt.polar(spiral_p[i],spiral_a[i],'8')
    
plt.polar(spiral_p,
          spiral_a,'--',
          color='cyan',
          alpha=0.5)

plt.axis('off')
plt.savefig('spiral plt.polar().pdf')
plt.show()
```

##### euler's formula plt.ploar

Euler' Formula is : 
$$ m \cdot e^{1j \cdot k} $$

```python
# create euler's number
eul_f_p = np.pi/4
eul_f_m = .5

eul_f_eulr = eul_f_m*np.exp(1j*eul_f_p)

# now extract magnitude and phase
eul_f_mag = np.abs(eul_f_eulr)
eul_f_ang = np.angle(eul_f_eulr)

# then plot
plt.polar([0,eul_f_ang],[0,eul_f_mag],'b',linewidth=3)
#            angle       magnitude 
plt.show()
```

## asteroid radial curve

$$ x=a \cdot cos (t)^3$$
$$ y=a \cdot cos (t)^3$$
$$ a \quad \text{angle is from 0  to } 2\cdot \pi  $$
$$ t \quad \text{time range from negative -15 to positive +15} $$

```python
ast_a=np.linspace(0,2*np.pi,100)
ast_t=np.linspace(-15,15,100)

ast_x=ast_a*np.cos(ast_t)**3
ast_y=ast_a*np.sin(ast_t)**3

# for i in range(100):
#     plt.plot(ast_x[i],ast_y[i],'s',alpha=1)
plt.plot(ast_x,ast_y,'-',color='brown',linewidth=5)
plt.axis('off')
plt.savefig('asteroid radial curve.pdf')
plt.show()
```

```python
ast_t1=np.linspace(-15,15,100)

fig,ax=plt.subplots(9,1,figsize=(2.5,18))

for i in range(9):


    ax[i].plot(      ast_t1**i*np.cos(ast_t)**3,
                     ast_t1**i*np.sin(ast_t)**3,
                     '-',
                     color='black',
                     linewidth=1)
    ax[i].axis('off')
plt.show()
```

```python
ast_t=np.linspace(-15,15,100)
row,col=np.indices((3,3))
fig,ax=plt.subplots(3,3,figsize=(15,10))
for i in range(9):
    ast_a=ast_t**i
    ast_x=ast_a*np.cos(ast_t)**3
    ast_y=ast_a*np.sin(ast_t)**3
    r=row.ravel()[i]
    c=col.ravel()[i]
    ax[r,c].plot(ast_x,
                 ast_y,
                 'blue')
    
    ax[r,c].axis('off')
    ax[r,c].axis('square')
plt.savefig('ss.pdf')
plt.show()
```

## rose curves

$$ x=cos(kt) \cdot cos(t)$$
$$ y=cos(kt) \cdot sin(t)$$
k even 2times quantity of the pedals $$ . $$
k odd  1times quantity of the pedals

## logarithm spiral

$$ x=cos(t)e^{t/k} $$
$$ x=sin(t)e^{t/k} $$
in order to create a declining effect we can set the k to a negative number 
as k increases, e's exponent gets smaller
decline 
$$ e^{-t} $$
wave
$$ cos(t) $$

```python
l_spir_t=np.linspace(0,10*np.pi,1234)
l_spir_k=-3
l_spir_x=np.cos(l_spir_t)*np.exp(l_spir_t/l_spir_k)
l_spir_y=np.sin(l_spir_t)*np.exp(l_spir_t/l_spir_k)

plt.plot(l_spir_x,l_spir_y,linewidth=5,color='green')
plt.axis('off')
plt.axis('square')
plt.savefig('d.pdf')
plt.show()
```

```python
l_spir_n=150

l_spir_k1=np.linspace( -6, -2,l_spir_n )

for i in range(l_spir_n):
    
    l_spir_x=np.cos(l_spir_t)*np.exp(l_spir_t/l_spir_k1[i])
    l_spir_y=np.sin(l_spir_t)*np.exp(l_spir_t/l_spir_k1[i])
    
    plt.plot(l_spir_x,l_spir_y,linewidth=1,color=[i/l_spir_n,
                                                  i/l_spir_n,
                                                  0])
plt.axis('off')
plt.axis('square')
plt.savefig('d.pdf')
plt.show()
```

```python
l_spir_n=25
l_spir_t1=np.linspace(0,10*np.pi,l_spir_n)
l_spir_k1=np.linspace( -6, -2,l_spir_n )

for i in range(l_spir_n):
    
    l_spir_x=np.cos(l_spir_t1)*np.exp(l_spir_t1/l_spir_k1[i])
    l_spir_y=np.sin(l_spir_t1)*np.exp(l_spir_t1/l_spir_k1[i])
    
    plt.plot(l_spir_x[i],
             l_spir_y[i],
             'o')

l_spir_x=np.cos(l_spir_t1)*np.exp(l_spir_t1/l_spir_k1)
l_spir_y=np.sin(l_spir_t1)*np.exp(l_spir_t1/l_spir_k1)
plt.plot(l_spir_x,
         l_spir_y)
plt.axis()
plt.axis('square')
plt.savefig('d.pdf')
plt.show()
```

## logistic map

$$ x_n+1=rx_n(1-x_n) $$

```python
# %matplotlib notebook
```

```python
l_map_r=3
l_map_n=500

l_map_x=0.5+np.zeros(l_map_n)

for n in range(l_map_n-1):
    l_map_x[n+1]= l_map_r*l_map_x[n]*(1-l_map_x[n])
    
plt.plot(l_map_x,'s-')

plt.show()
```

#Sine to find side measures

### side measures

```python
import math

# make the unknown value zero
angle = 57
opposite = 4
hypotenuse = 0

if opposite==0:
    unknown = "opposite side"
    answer = hypotenuse * math.sin(math.radians(angle))

if hypotenuse==0:
    unknown = "hypotenuse"
    answer = opposite/math.sin(math.radians(angle))

print("The ", unknown, " is ", answer)

```

#Cosine to find side measures

```python
import math

# make the unknown value zero
angle = 57
adjacent = 3
hypotenuse = 0

if adjacent==0:
    unknown = "adjacent side"
    answer = hypotenuse * math.cos(math.radians(angle))

if hypotenuse==0:
    unknown = "hypotenuse"
    answer = adjacent/math.cos(math.radians(angle))

print("The ", unknown, " is ", answer)
```

#Pythagorean trig

sin<sup>2</sup>(x) + cos<sup>2</sup>(x) = 1

```python
import math

for a in range(70):
    angle = 0.1*a
    y = math.sin(angle)**2
    x = math.cos(angle)**2
    r = x + y
    print(x, " + ", y, " = ", r )
```

#Radians and Degrees

$\frac{2π \;radians}{360 \;degrees}$ = $\frac{radians}{degrees}$

### sin transformation

#Sine wave transformations

```python
import matplotlib.pyplot as plt
from IPython import display
import numpy as np
import math
%matplotlib inline
from ipywidgets import interactive

# Graph dimensions
xmin = -2
xmax = 14
ymin = -5
ymax = 5
points = 10*(xmax-xmin)
x = np.linspace(xmin, xmax, points)

def graph(a,b,c):
    fig, ax = plt.subplots()
    eq = 'sin('
    eq2 = 'x) + '
    plt.title(f'{a}{eq}{b}{eq2}{c}')
    plt.axis([xmin,xmax, ymin,ymax])
    plt.plot([xmin, xmax],[0,0],'b')
    plt.plot([0,0],[ymin, ymax],'b')
    y = a*np.sin(b*x)+c
    plt.plot(x,y,'r')
    plt.show()

ig = interactive(graph, a=(-5,5), b=(-5,5), c=(-5,5))
ig
```

#Interacting sine waves
Notice the patterns when you add two sine functions ("additive interference")

```python
import matplotlib.pyplot as plt
from IPython import display
import numpy as np
import math

# Graph dimensions
xmin = -2
xmax = 14
ymin = -2
ymax = 2
points = 10*(xmax-xmin)
x = np.linspace(xmin, xmax, points)

fig, ax = plt.subplots()
plt.axis([xmin,xmax, ymin,ymax])
plt.plot([xmin, xmax],[0,0],'k')
plt.plot([0,0],[ymin, ymax],'k')

# change these
y1 = np.sin(200*x)
y2 = np.sin(300*x)
y3 = y1 + y2

plt.plot(x,y1,'b')
plt.plot(x,y2,'g')
plt.plot(x,y3,'r')
plt.show()
```

## Triangle from points
distance formula three times<br>
law of cosines twice<br>
subtraction



```python
def triangle(x1,y1,x2,y2,x3,y3):
    import math
    # get side a
    a = math.sqrt((x2-x1)**2 + (y2-y1)**2)
    # get side b
    b = math.sqrt((x3-x2)**2 + (y3-y2)**2)
    # get side c
    c = math.sqrt((x3-x1)**2 + (y3-y1)**2)
    # get angle a
    cos_a = (a**2 - b**2 - c**2)/(-2*b*c)
    a_radians = math.acos(cos_a)
    a_degrees = math.degrees(a_radians)
    # get angle b
    cos_b = (b**2 - a**2 - c**2)/(-2*a*c)
    b_radians = math.acos(cos_b)
    b_degrees = math.degrees(b_radians)
    # get angle c
    c_radians = math.pi - (a_radians + b_radians)
    c_degrees = 180 - (a_degrees + b_degrees)
    # get area
    area = 0.5*b*c*math.sin(a_radians)
    # output
    print("Angle A = ", round(a_degrees,1))
    print("Side A = ", round(a,1))
    print("Angle B = ", round(b_degrees,1))
    print("Side B = ", round(b,1))
    print("Angle C = ", round(c_degrees,1))
    print("Side c = ", round(c,1))
    print("Area = ", round(area,1))

# Call the function
triangle(0,3,0,0,4,0)
```

### recursive sequence fibonacci

```python
def fibonacci(n):
   if n <= 1:
       return n
   else:
       return fibonacci(n - 1) + fibonacci(n - 2)
# Example usage
num_terms = 10
print("Fibonacci series up to 10 terms:")
for i in range(num_terms):
   print(fibonacci(i))
```

## Hours of daylight
<p>
source:
 https://web.archive.org/web/20180128172128/http://mathforum.org/library/drmath/view/56478.html



```python
import math

# latitude in degrees (up to 65)
L = 65

# day of the year
J = 173

# calculations in several steps
t1 = 0.9671396*math.tan(0.00860*(J-186))
p = math.asin(0.39795*math.cos(0.2163108 + 2*math.atan(t1)))
d1 = math.sin(0.8333*math.pi/180) + math.sin(L*math.pi/180) * math.sin(p)
d2 = math.cos(L*math.pi/180)*math.cos(p)
daylight = 24 - (24/math.pi)*math.acos(d1/d2)
print(round(daylight,1), " hours")
```

## Poisson Distribution

y = $\frac{\lambda^x}{x!}*e^{-\lambda}$<p>
where lambda ($\lambda$) is the expected value<p>
What is the probability that x happens?

```python
import math
import matplotlib.pyplot as plt
from IPython import display
import numpy as np

# expected value
lam = 4

# Graph dimensions
xmin = 1
xmax = 10 # could change
ymin = 0
ymax = 1

# create arrays to plot
x = []
y = []

for x_next in range(1,10):
    x.append(x_next)
    y.append((lam**x_next)/math.factorial(x_next)*math.e**(-lam))

fig, ax = plt.subplots()
plt.axis([xmin,xmax, ymin,ymax])
plt.plot([xmin, xmax],[0,0],'b')
plt.plot([0,0],[ymin, ymax],'b')

plt.plot(x,y,'r')

plt.show()
```

```python
# adding vectors
v1 = np.array([-1,2])
v2 = np.array([1,1])

v3a = v1+v2                # method 1 using addition operator
v3b = np.add(v1,v2)        # method 2 using np.add() method and pass two vectors 
                           # as arguments
v3c = np.zeros(2)          # method 3 using for loop to add element pair that share 
for i in range(0,2):       # the same index one at a time , 
    v3c[i] = v1[i] + v2[i]

print(v3a,v3b,v3c)
```

### visualization of th geometric 

```python
M_vs_geo= np.array([ [.5,1] ,
                     [1,.5] ])

V_vs_geo= np.array([1,2])

MV_vs_geo=M_vs_geo@V_vs_geo
fig,ax=plt.subplots()

plt.plot([0,V_vs_geo[0]],
         [0,V_vs_geo[1]],'b',
         linewidth=2,
         label='v')

plt.plot([0,MV_vs_geo[0]],
         [0,MV_vs_geo[1]],'r--',
         linewidth=2,
         label='MV-product of M &n V')

plt.axis('square')
plt.axis([-5,5,-5,5])
plt.legend()
plt.grid(alpha=0.2)
plt.text(-4,-2,s='When you multiply a vector(V) by a matrix(M)\nTHE product (MV) will rotate angle and \nchange the magnitude \neither shrink or expand the length of v ',
        color='red')
plt.text(-4,1,s='matrix\n[ [.5,1]\n [1,.5] ]\n\nvector[1,1]',color='blue',fontsize=10)
ax.set_xticks(np.arange(-3.5,4.6,1))
plt.plot(2.5,2,'r*',ms=10)
plt.text(2.8,2,s='MV(2.5,2)',color='red')

plt.show()

# When you multiply a vector(V) by a matrix(M)
#  THE product (MV) will rotate angle and change the magnitude 
#                      000 shrink and expand the length 
```

```python
# part 2: 6x6 symmetric matrix. compute all pairwise dot products of eigenvectors
# the matrix
m = 6
M = np.random.randn(m,m)
M_mul_t = M@M.T  #（ symmetic matrix =A@AT

# its eigendecomposition
d6,V6 = np.linalg.eig(M_mul_t)

dps = np.zeros((m,m))

# compare dot product between each col pairs - which 
# means two consecutive columns in the eigenvector 
for i in range(0,m):
  for j in range(0,m):
    dps[i,j] = np.dot(V6[:,i],V6[:,j])
np.round(dps,2)   
```

```python
# the idea of Eigenddecomposition 
#  is to extract eigenvalues and eigenvectors from a square matrix 
```

$$\text{Matrix A}   - \lambda \cdot \text{IdentityMatrix} $$ 
$$\begin{bmatrix}
    1 & 1  \\
    4 & 1  \\ 
\end{bmatrix} - \begin{bmatrix}
    \lambda & 0 \\
    0 & \lambda \\
\end{bmatrix} = \begin{bmatrix}
    1-\lambda & 1 \\
    4 & 1-\lambda \\
\end{bmatrix}
$$
$$\text{Determinant of the Matrix of Difference between  A and } \lambda \cdot   \text{IdentityMatrix is zer} o$$
$$(1-\lambda)(1-\lambda)-4\cdot1 $$ 


$$ \lambda^2-2\lambda-3=0 $$
$$\lambda_1=-1 \quad \lambda_2=3 $$

$$ $$
$$----------solving---eigenvector ----$$
$$ $$
plug in lambda =3
$$\text{Matrix A}   - \lambda \cdot \text{IdentityMatrix} $$ 
$$\begin{bmatrix}
    1 & 1  \\
    4 & 1  \\ 
\end{bmatrix} - \begin{bmatrix}
    3 & 0 \\
    0 & 3 \\
\end{bmatrix} = \begin{bmatrix}
    -2 & 1 \\
    4 & -2 \\
\end{bmatrix}= \text{row1}\cdot2+\text{row2} \begin{bmatrix}
    -2 & 1 \\
    0 &  0 \\
\end{bmatrix} \begin{bmatrix}
    x_1  \\
    x_2  \\
\end{bmatrix}
$$
$$-2x_1+x_2=0  \quad \rightarrow x_2=2x_1 $$
let's choose x_1=1 then x_2 is 2 thus the eigenvector ast eigenvalue lambda=3 is

$$\begin{bmatrix}
    x_1 \\
    x_2 \\
\end{bmatrix} = \begin{bmatrix}
    1 \\
    2 \\
\end{bmatrix}$$

plug in lambda =-1
$$\text{Matrix A}   - \lambda \cdot \text{IdentityMatrix} $$ 
$$\begin{bmatrix}
    1 & 1  \\
    4 & 1  \\ 
\end{bmatrix} - \begin{bmatrix}
    -1 & 0 \\
    0 & -1 \\
\end{bmatrix} = \begin{bmatrix}
    2 & 1 \\
    4 & 2 \\
\end{bmatrix}=\text{row1}\cdot2-\text{row2} \begin{bmatrix}
    2 & 1 \\
    0 & 0 \\
\end{bmatrix} \begin{bmatrix}
    x_1  \\
    x_2  \\
\end{bmatrix}
$$
$$2x_1+x_2=0 \quad \rightarrow x_1= - \frac{1}{2} \cdot x_2$$
let's choose x_1=1 then x_2 is -2 thus the eigenvector ast eigenvalue lambda=-1 is
$$\begin{bmatrix}
    x_1 \\
    x_2 \\
\end{bmatrix} = \begin{bmatrix}
    1 \\
    -2 \\
\end{bmatrix}$$
summary :
as long as the second element in the vector is twice the first
then its an eigenvector

```python
# step1 
# solving the eigen values 
# first subtrct lambdaidentity matrix from the original matrix 
# thus forming a polynomial equation 
# afterward , solve the equation  
```

```python
# step2
# plug in eigenvalue one by one  back into the quation 
# step3
# reduce rows into row eschelon and then perform matrix vector multiplication 
# solving the equation by assigning 1 to the first element 
# solution verification : as long as the proportion between elements within 
# the vector is corrent then its an eigenvector 
```

```python
# eigenvector is constant c times the solution 
```

```python
# the determinant of the matrix must equal to zero , 
#   therefore , it solves the eigen values 
#     afterwards using the eigen values to solve eigen vectors 
#       
```

```python
# matrix multiplied by a vector - the product is another vector 
# rotate and change the magnitude of the original vector 
#  with a few exceptions that won't rotate but only shrink or expand 
#     --another 180 degree the opposite direction 

# ------ therefore, if the product vector of the original vector and 
# ------ matrix is pointing the same direction as the original vector , 
# ------ the matrix only scales the original vector.

```

```python
M_Eig=np.random.randint(-5,6, ( 5 , 5 ) )  # must be a square matrxi 
                          #     row and 
                          #     col
M_Eig=M_Eig.T
eigvals,eigvecs=np.linalg.eig(M_Eig)
# eigvals.astype(float)  #------three ways to convert non-numeric to numeric
# eigvecs.astype(float)  # 1.d.astype(float) 2pd.to_numeric(df,sereis ) 3d.real

fig,ax=plt.subplots(1,4)
ax[0].imshow(M_Eig)
ax[0].set_title('Matrix A')

ax[1].imshow(M_Eig.T)
ax[1].set_title('Matrix AT')

ax[2].imshow(np.diag(eigvals.real)) # eigvals are an array of elements therefore 
ax[2].set_title('AT eigvals')       # its necessary to convert it using np.diag to 
                                    # propogate those elements along diagonal axis
ax[3].imshow(eigvecs.real)
ax[3].set_title('AT eigvecs')
```

### application statistics (correlation coefficient )

#### def

```python
# correlation coefficient 
#  is a popular way to dertermine whether there's a 
#   correlation between two variables 

#========================== def =================================

# A measure of the interdependence of two random variables 
# that ranges in value from -1 to +1,
#                              indicating 
                                      # perfect negative correlation at -1, 
                                      # absence of correlation at zero, and 
                                      # perfect positive correlation at +1. 
        #     Also called coefficient of correlation.
```

$$ r_{xy}=\frac{\sum_{i=1}^n(x_i- \bar{x}) \cdot (y_i- \bar{y})} {\sqrt{\sum_{i=1}^n(x_i- \bar{x})^2 \cdot \sum_{i=1}^n(y_i- \bar{y})^2}} $$
numerator sum up the difference between individual x and avgx times the one y and avgy
$$ $$
denominator difference 's square and the square root time y counterpart 
$$ $$
square and the square root is to elimitae positive and negative deviation neutralization

$$-------------numerator------------------ $$
$$\text{-------difference between x[index i ] and mean of  x collection \quad} x_i- \bar{x} $$ 
$$\text{-------difference between y[index i ] and mean of  x collection \quad} y_i- \bar{y} $$ 
$$\text{-------product of differences } \quad (x_i- \bar{x}) \cdot (y_i- \bar{y}) \quad \quad \quad\quad\quad\quad\quad\quad\quad$$
$$\text{-------summation of products } \sum_{i=1}^n(x_i- \bar{x}) \cdot (y_i- \bar{y})\quad\quad\quad\quad\quad\quad\quad\quad $$

$$-------------denominator------------------ $$
standardizing - absolute ||
$$ \sqrt{n^2} =|n| \text{to make sure that the proportion is <=1 or -1}$$
$$-------------overview------------------ $$
sum(deviation squares from mean) over abs|product |
all data points' deviation from the means(x,y) 
$$-------------best fit line------------------ $$
line of best fit is also called the least square regression line 
regression - fall back -regret.regression 
$$ $$ minimize the distance from y to mean y and x to mean 

#### formula (mannual calculation)

```python
mikes_course  =[4,5,8,2,9,9,1,3,7]
life_happiness=[6,7,9,3,9,3,1,6,7]
m_d=mikes_course - np.mean(mikes_course)
l_d=life_happiness -np.mean(life_happiness)
```

```python
numer=np.dot(m_d,l_d) # sum of (difference x y's product)
```

```python
denom=np.sqrt(np.dot(m_d,m_d))*np.sqrt(np.dot(l_d,l_d)) # sqrt (product of l m -sum)
```

```python
cour_r1=numer/denom
cour_r2=np.corrcoef(mikes_course,life_happiness)
```

```python
v_d_pa=np.arange(10,19,3)  #
v_d_pb=np.arange(5,14,3)   # its necesssary to check the lenghth of the 

#  two arrays to make sure that they match prior to the product 
# set an initial value and the use for loop to update 
#  the each multiplication's product by add it and then 
#   assign to the variable , thus each for loop it updates 
#     and assumes teh sum of previous value of itsself
#        and the addition 

# 0-sanity check ------
dp1=0
for i in range(0,len(v_d_pa)):
    if len(v_d_pa)==len(v_d_pb):
        dp1=int(dp1+v_d_pa[i]*v_d_pb[i])
    else:
        raise ValueError ('The lens of the two vectors do not match')
```

#### def

$$\alpha = a \cdot b =<a,b>=a^Tb=\sum_{i=1}^n a_ib_i $$
outcome is a scalar which means it produces a single number 
$$ \sum_{i=1}^n a_ib_i $$
operational procedure of dot product
$$ $$ 
first loop through a and b using index i to gain access to to 
both a and b vector that $$ $$ share the same index -after multiplication of 
ai and bi $$ $$ then sum up the product of the pairs with identical index 

#### why

###################polar coordinates###################
$$ \overrightarrow{ a } \cdot \overrightarrow{ b } = |a||b|cos\theta$$\
formula rewrite 
$$ \overrightarrow{ a } \cdot \overrightarrow{ b } = (|a|\cdot cos\theta) \cdot |b|$$\
projection of a on to the b vector so that they are pointing at the same direction 
$$ (|a|\cdot cos\theta) \cdot |b|=(|b|\cdot cos\theta) \cdot |a| $$
commutativity - commutative property 
how much of one vector is lying on the other vector 
we are only mutiply two vectors in the manner that one is 
proportional to its own length that is in the same direction the otherone
is pointing 
$$ $$
application - angle - distance and force 
$$ $$
dot product of force and distance at angle is the work 
$$ \text{W=F} \cdot \text{D} $$ 
$$ \text{Work = Force} \cdot \text{Distance} $$
special case
$  \quad cos\theta  =0^o $ means $cos =1$ maximum $ \quad cos\theta =90^o $ means cos=0 minimum 
$$ $$ 
###################cartisean coordinates###################
$$ \overrightarrow{a}=a_x \hat{i}  +  a_y \hat{j} + a_z \hat{k}$$
$$ \overrightarrow{b}=b_x \hat{i}  +  b_y \hat{j} + b_z \hat{k}$$
$$ \overrightarrow{ a } \cdot \overrightarrow{ b } =a_xb_x+a_yb_y+a_zb_z$$
################ cartisean -polar conversion ###############
using pythagrean theorem to calculate the magnitude of the
vector one by one 
example
$$ \overrightarrow{a}=3\hat{i}+2\hat{j} \quad \overrightarrow{b}=\hat{i}+4\hat{j} $$
$$ |\overrightarrow{a}|= \sqrt{3^2+2^2}$$
=33.69 (b_angle) 
$$ |\overrightarrow{b}|= \sqrt{1^2+4^2}$$
=75.96 (b_angle)
$$ |\overrightarrow{a}|\cdot|\overrightarrow{b}|\cdot ({\theta b -\theta a}) $$
#################### proof -dot formula ####################
derivation 
$$|\overrightarrow{a}|\cdot|\overrightarrow{b}|=(a_x\hat{i}+a_y\hat{j}+a_z\hat{k})\cdot(b_x\hat{i}+b_y\hat{j}+b_z\hat{k})$$
expansion of the multiplication
$$=a_xb_x\hat{i}\cdot\hat{i}(1)+a_xb_y\hat{i}\cdot\hat{j}(0)+a_xb_z\hat{j}\cdot\hat{k}(0)$$ 
$$+a_yb_x\hat{j}\cdot\hat{i}(0)+a_yb_y\hat{j} \cdot \hat{j}(1)+a_yb_z\hat{j}\cdot\hat{k}(0)$$
$$+a_zb_x\hat{k}\cdot\hat{i}(1)+a_zb_y\hat{k} \cdot \hat{j}(0)+a_zb_z\hat{k}\cdot\hat{k}(1)$$ 
$$ $$
comments - ii = means zero angel -that's cos 0 =1  ij-= means 90   angle -that's cos 90=0      

simplification
$$=a_xb_x\hat{i}\cdot\hat{i}(1)++a_yb_y\hat{j}\cdot\hat{j}(1)+a_zb_z
\hat{k}\cdot\hat{k}(1)$$ 
further simplification by removing the coefficient 1
$$=a_xb_x+a_yb_y+a_zb_z$$

```python
# number of data points
n_outerprod = 10

# data
data1 = np.arange(0,n) + np.random.randn(n_outerprod)
data2 = np.arange(0,n) + np.random.randn(n_outerprod)

data1mean=np.mean(data1)
data2mean=np.mean(data2)

# compute correlation
numer_corrcoef = np.dot((data1-data1mean),(data2-data2mean))
denom_corrcoef = np.sqrt( np.dot((data1-data1mean),(data1-data1mean)) ) * np.sqrt(np.dot((data2-data2mean),(data2-data2mean)))
corrcoef_1 = numer_corrcoef/denom_corrcoef

# confirm with numpy function
corrcoef_2 = np.corrcoef(data1,data2)

print(corrcoef_1)
print(corrcoef_2)
corrcoef_1==corrcoef_2[0][1]
```

## inverse

#### def

```python
# why inverse of matrix is need to solve varibales in a matrix of equation systems 
#  Ax=B to solve x -multiply both sides with A inverse A-1 1/A = the left side is ruduce to 1X=B/A
#    THUS X=B/A  to expand the theory into matrix 
#      Mcoefficient = Maugmented  invisible variables are existing -  Mcoe*Mcoe-inv(invisible variable =Maug*Mcoe)#   
#        to produce identity matrix - thus the equation systems are solved 
```

#### why

$$3x=1 \quad \Rightarrow $$
multiply both sides by a factor of $ 3^{-1}$
$$ 3^{-1} \cdot 3x = 3^{-1} \cdot 1  $$
simplify variable's coefficients 
$$1x=3^{-1} $$
further simplify 
$$x=3^{-1} $$

original equation systems 
$$2x + 6y + 6z = 8 $$
$$2x + 7y + 6z = 10 $$
$$2x + 7y + 7z = 9  $$
step 1 form a matrix by extracting coefficients -A 
$$\begin{bmatrix}
2 & 6 & 6  \\
2 & 7 & 6  \\
2 & 7 & 7 
\end{bmatrix}$$
step 2 form a matrix using augmented values on the right side of the quation -B
$$\begin{bmatrix}
8   \\
10  \\
9  
\end{bmatrix}$$
step 3 inverse of the matrix A$^T$
$$\begin{bmatrix}
1 & -1 & 1  \\
0 & 2 & -1  \\
2 & 3 & 0 
\end{bmatrix}$$
step 4 multiply both sides by inverse matrix  A$^T$
left side is an identity matrix 
$$\begin{bmatrix}
1 & 0 & 0  \\
0 & 1 & 0  \\
0 & 0 & 1 
\end{bmatrix}$$  
right side is the answer
$$\begin{bmatrix}
1   \\
2  \\
-1  
\end{bmatrix}$$

#### pseudo inverse

```python
# a matrix is invertible if its 
# 1 square     (
# 2 full rank  (unique coulumn values - each columns ' values must be unique)
# - otherwise, use the pseudoinverse !
```

```python
pin_m=5
pin_A=np.random.randint(-pin_m,pin_m+1,(pin_m,pin_m))
pin_A[:,0]=pin_A[:,1] # SET THE firt column equal the second column 
                      # mannually change the column values to create duplicates 

a_pinv=np.linalg.pinv(pin_A)

a_pinv

fig,ax=plt.subplots(1,3)

ax[0].imshow(pin_A)
ax[0].set_title('A')

ax[1].imshow(a_pinv)
ax[1].set_title('$A^T-pseudo$')

ax[2].imshow(np.matmul(pin_A,a_pinv))
ax[2].set_title('$A*A^T$')
np.round(pin_A,2)
```

```python
pin_m=5
pin_A=np.random.randn(pin_m,pin_m+2) #nonsquare row and column numbers are diff
pin_A[:,0]=pin_A[:,1] # SET THE firt column equal the second column 

a_pinv=np.linalg.pinv(pin_A)

a_pinv

fig,ax=plt.subplots(1,3)

ax[0].imshow(pin_A)
ax[0].set_title('A-Gaussian\n Distribution')

ax[1].imshow(a_pinv)
ax[1].set_title('$A^T-pseudo$')

ax[2].imshow(np.matmul(pin_A,a_pinv))
ax[2].set_title('$A*A^T$')
```

Add [A] + [B] = [C]

```python
# add corresponding numbers
c11 = a11 + b11
c12 = a12 + b12
c21 = a21 + b21
c22 = a22 + b22

print(c11,c12)
print(c21,c22)
```

Multiply [A] * [B] = [C]

```python
import numpy as np

# enter all values
a11=1
a12=2
a21=3
a22=4

b11=5
b12=6
b21=7
b22=8

# The actual steps
c11 = a11*b11 + a12*b21
c12 = a11*b12 + a12*b22
c21 = a21*b11 + a22*b21
c22 = a21*b12 + a22*b22

print("Multiplying Matrix A * Matrix B:")
print(c11,c12)
print(c21,c22)

# numpy
print("\n same matrix:")
A = np.array([[a11, a12], [a21, a22]])
B = np.array([[b11,b12], [b21,b22]])
C = A.dot(B)
print(C)

# np.matrix
print("\n np.matrix:")
E = np.matrix([[a11, a12], [a21, a22]])
F = np.matrix([[b11,b12], [b21,b22]])
G = E.dot(F)
print(G)

```

Converting matrix and array objects in numpy

```python
import numpy as np
from sympy import *

# Create an array
arr = np.array([[14, 0, 11, 3], [22, 25, 4, 7], [-12, -34, -3, -4]])

# Create a matrix from that array
M = Matrix(arr)
print("Matrix : {} ".format(M))

# could start by creating a matrix instead
# M = Matrix([[14, 0, 11, 3], [22, 23, 4, 7], [-12, -34, -3, -4]])

# convert that matrix back to an array
A = np.array(M)
print("\n same matrix:")
print(A)

# Solve using rref in the matrix
print("\n answers:")
M_rref = M.rref()
x = M_rref[0][3]
y = M_rref[0][7]
z = M_rref[0][11]
print("x = ", x)
print("y = ", y)
print("z = ", z)

print("\n row echelon form of matrix M: \n {}".format(M_rref))
```

points: (x1, y1), (x2, y2), (x3, y3)<p>
Standard equation: Ax<sup>2</sup> + Bx + C = y<br>
first point: A(x1)<sup>2</sup> + B(x1) + C = y1<br>
Second point: A(x2)<sup>2</sup> + B(x2) + C = y2<br>
Third point: A(x3)<sup>2</sup> + B(x3) + C = y3<p>
Matrix:<br>
$\begin{bmatrix} (x1)^2 & x1 & 1 & y1
\\ (x2)^2 & x2 & 1 & y2
\\ (x3)^2 & x3 & 1 & y3  \end{bmatrix}$

### ---solving matrix

### M.rref -solving matrix reduced row eschalon form

```python
import numpy as np
from sympy import *

# points
x1=-2
y1=0
x2=-3

y2=0
x3=-2.5
y3=-.25


M = Matrix([[x1**2,x1,1,y1],[x2**2,x2,1,y2],[x3**2,x3,1,y3]])

m_rref = M.rref()  # Reduced row echelon form

# answers
a = round(m_rref[0][3])
b = round(m_rref[0][7])
c = round(m_rref[0][11])
print("a = ", a)
print("b = ", b)
print("c = ", c)

print("y = ", a,"x**2 + ", b, "x + ", c)
print("\n row echelon form of matrix M: \n {}".format(m_rref))

```

### Inverse matrix to solve a system of equations

```python
# essene of solving  matrix equation system 

#  step 1 . using coefficients of variables to construct a matrix 
#  step 2 . using results to form an augmented matrix 

#  ---
#  step3 . solve the inverse of the matrix composed of coefficients 
#  step3 . soluton = inverse of the coefficient matrix  * matrix of augmented values 

#---area to be updated 

import numpy as np
from sympy import *

# use arrays instead of matrices
               #ax  by  cz 
               #123 123 123

a1 = np.array([[14, 0, 11], 
               [22, 23, 4], 
               [-12, -34, -3]])

a2 = np.array([[3],     # ax1 + by1 +cz1 =3 
               [7],     # ax2 + by2 +cz2 =3 
               [-4]])   # ax3 + by3 +cz3 =3 


# why matrix of coefficient's columnes must be identical with 
# augmented values matrix\s rows ? 
#  reason is becuase there are n variable and the correponding x augmented value - 
# Simply put it takes n equations to solve n variable matrix .


m_inv = np.linalg.inv(a1)  
m_dot_a = np.around(np.dot(m_inv,a1))
m_dot = np.dot(m_inv,a2)

print(f"m_inv\n{m_inv}")
print('\n')
print(f"m_dot_a\n{m_dot_a}")
print('\n')
print(f"m_dot\n{m_dot}")
print('\n')

print("\n compare:")
print(445/1358)
print(10/679)
print(-14/97)

print(m_dot_a)
```

Another inverse matrix example

```python
import numpy as np
from sympy import *

# points
x1=-2
y1=0
x2=-3

y2=0
x3=-2.5
y3=-.25

# rref method
M = Matrix([[x1**2,x1,1,y1],[x2**2,x2,1,y2],[x3**2,x3,1,y3]])
m_rref = M.rref()

# answers
a = round(m_rref[0][3])
b = round(m_rref[0][7])
c = round(m_rref[0][11])
print("a = ", a)
print("b = ", b)
print("c = ", c)

# inverse matrix method
arr_3x3 = np.array([[x1**2,x1,1],[x2**2,x2,1],[x3**2,x3,1]])
arr_3x1 = np.array([[y1],[y2],[y3]])
arr_inv = np.linalg.inv(arr_3x3)
identity = np.around(np.dot(arr_inv,arr_3x3))
arr_values = np.dot(arr_inv,arr_3x1)
print(identity)
print("answers")
print(arr_values)
print("a = ", round(arr_values.item(0)))
print("b = ", round(arr_values.item(1)))
print("c = ", round(arr_values.item(2)))
```

## Eigenvalues and Eigenvectors

#### Eigenddecomposition

```python
# the idea of Eigenddecomposition 
#  is to extract eigenvalues and eigenvectors from a square matrix 
# important Eigenddecomposition is defined 
# only for square matrices m x m 
# =======singular value decomposition works 
# for any m x n matrix 
```

#### Eigen v pairs

##### solving mannually

$$A\cdot \overrightarrow{X}=\lambda \cdot \overrightarrow{X} $$

$$A\cdot \overrightarrow{X} - \lambda \cdot \overrightarrow{X} =\overrightarrow{0}$$ 

$$\overrightarrow{X}\cdot (A- \lambda \cdot \text{IdentityMatrix}  ) =\overrightarrow{0}$$ 
characteristics of identity matrix $$ $$ is that it's propogated with ones along the diagonal axis 
and zeros in both the upper and lower triangular areas 
$$ $$
$$----------solving---eigenvalues ----$$
$$ $$
to solve eigenvalues , first ignore the eigen vector 
$$A- \lambda \cdot \text{IdentityMatrix}   =\overrightarrow{0}$$ 
solving polynomials
$$ $$ example:

$$\text{Matrix A}   - \lambda \cdot \text{IdentityMatrix} $$ 
$$\begin{bmatrix}
    1 & 1  \\
    4 & 1  \\ 
\end{bmatrix} - \begin{bmatrix}
    \lambda & 0 \\
    0 & \lambda \\
\end{bmatrix} = \begin{bmatrix}
    1-\lambda & 1 \\
    4 & 1-\lambda \\
\end{bmatrix}
$$
$$\text{Determinant of the Matrix of Difference between  A and } \lambda \cdot   \text{IdentityMatrix is zer} o$$
$$(1-\lambda)(1-\lambda)-4\cdot1 $$ 


$$ \lambda^2-2\lambda-3=0 $$
$$\lambda_1=-1 \quad \lambda_2=3 $$

$$ $$
$$----------solving---eigenvector ----$$
$$ $$
plug in lambda =3
$$\text{Matrix A}   - \lambda \cdot \text{IdentityMatrix} $$ 
$$\begin{bmatrix}
    1 & 1  \\
    4 & 1  \\ 
\end{bmatrix} - \begin{bmatrix}
    3 & 0 \\
    0 & 3 \\
\end{bmatrix} = \begin{bmatrix}
    -2 & 1 \\
    4 & -2 \\
\end{bmatrix}= \text{row1}\cdot2+\text{row2} \begin{bmatrix}
    -2 & 1 \\
    0 &  0 \\
\end{bmatrix} \begin{bmatrix}
    x_1  \\
    x_2  \\
\end{bmatrix}
$$
$$-2x_1+x_2=0  \quad \rightarrow x_2=2x_1 $$
let's choose x_1=1 then x_2 is 2 thus the eigenvector ast eigenvalue lambda=3 is

$$\begin{bmatrix}
    x_1 \\
    x_2 \\
\end{bmatrix} = \begin{bmatrix}
    1 \\
    2 \\
\end{bmatrix}$$

plug in lambda =-1
$$\text{Matrix A}   - \lambda \cdot \text{IdentityMatrix} $$ 
$$\begin{bmatrix}
    1 & 1  \\
    4 & 1  \\ 
\end{bmatrix} - \begin{bmatrix}
    -1 & 0 \\
    0 & -1 \\
\end{bmatrix} = \begin{bmatrix}
    2 & 1 \\
    4 & 2 \\
\end{bmatrix}=\text{row1}\cdot2-\text{row2} \begin{bmatrix}
    2 & 1 \\
    0 & 0 \\
\end{bmatrix} \begin{bmatrix}
    x_1  \\
    x_2  \\
\end{bmatrix}
$$
$$2x_1+x_2=0 \quad \rightarrow x_1= - \frac{1}{2} \cdot x_2$$
let's choose x_1=1 then x_2 is -2 thus the eigenvector ast eigenvalue lambda=-1 is
$$\begin{bmatrix}
    x_1 \\
    x_2 \\
\end{bmatrix} = \begin{bmatrix}
    1 \\
    -2 \\
\end{bmatrix}$$
summary :
as long as the second element in the vector is twice the first
then its an eigenvector

```python
# step1 
# solving the eigen values 
# first subtrct lambdaidentity matrix from the original matrix 
# thus forming a polynomial equation 
# afterward , solve the equation  
```

```python
# step2
# plug in eigenvalue one by one  back into the quation 
# step3
# reduce rows into row eschelon and then perform matrix vector multiplication 
# solving the equation by assigning 1 to the first element 
# solution verification : as long as the proportion between elements within 
# the vector is corrent then its an eigenvector 
```

```python
# eigenvector is constant c times the solution 
```

```python
# the determinant of the matrix must equal to zero , 
#   therefore , it solves the eigen values 
#     afterwards using the eigen values to solve eigen vectors 
```

```python
# matrix multiplied by a vector - the product is another vector 
# rotate and change the magnitude of the original vector 
#  with a few exceptions that won't rotate but only shrink or expand 
#     --another 180 degree the opposite direction 

# ------ therefore, if the product vector of the original vector and 
# ------ matrix is pointing the same direction as the original vector , 
# ------ the matrix only scales the original vector.
```

### piecewise function 

#### def

$$ Function = \begin{cases}
Formula1 & Condition1 \\
Formula2 & Condition2 \\
Formula3 & Condition3 
\end{cases}$$

$$ f(x) = \begin{cases}
\quad 0 & \text{if} \quad x <= 0 \\
-2x & \text{if} \quad x > 0 \\
\end{cases}$$

```python
# how to write piecefunction in latex 

# \begin{cases }
#------------
# \end{cases}
```

```python
# In mathematics, a piecewise-defined function 
#  (also called a piecewise function, a hybrid function, or definition by cases) 
 #   is a function defined by multiple sub-functions, 
    #  where each sub-function applies to a different interval in the domain.
      #      -----------------              --------------------------------

# a piecewise function 
  # is a function that can't be expressed using one single expression   
       #   instead it must be expressed using multiple expressions 
```

```python
p_wise_x=np.linspace(-10,10,50)

for i in p_wise_x :
    if i <=0:
        plt.plot(i,0,'o')
    else:
        plt.plot(i,-2*i,'+')
plt.text(-10,-20,s='o---f(x)=0 if x<=0 \n\n+---f(x)=-2x if x>0')

plt.plot([0,0],[-20,-0.2],'k--')
plt.plot([0,10],[0,0],'k--')
plt.title('piecewise functions')

plt.show()
```

$$ f(x) = \begin{cases}
0 & x<=0  \\
-2x &  0< x <10 \\
.1x^3 & x=>10
\end{cases}$$

```python
from sympy.abc import x 

piece1=0
piece2=-2*x
piece3=x**3/10

px=sympy.Piecewise( ( piece1 , x<0             ),
                    ( piece2 , (x>=0)&(x<10)   ),
                    ( piece3 , x>=10           ))

fxxf=sympy.lambdify(x,px)
xxpw=np.linspace(-12,12,50)
plt.plot(xxpw,fxxf(xxpw),color='blue',linewidth=1)

plt.title('piecewise function-lambdify(boundary -12,12)')

plt.show()
# piecewise function method in sympy syntax 
#                    each formula and condition pair must be encompased within a pair of parentheses 
#                    each pair must be separated with a comma 
#                        if two or more condtions exsit within one formula and condition pair , then 
#                          the each conditon must be wrapped within a pair of parenthesis 
```

### Limits - approaching from right or left<br>
Example function: y = $\frac{3x^2}{x^2 - 4}$


```python
x = -2
h = 0.0001

# for y = (3*x**2)/(x**2 - 4)

y_right1 = (3*((x+h)**2)) / ((x+h)**2 - 4)
y_left1 =  (3*((x-h)**2)) / ((x-h)**2 - 4)

print("y_right1 = ", round(y_right1,))
print("y_left1 = ", round(y_left1,))

if round(y_right1) != round(y_left1):
    print("Limit does not exist at x = ", x)
```

```python
x = 2
h = 0.0001

# for y = (3*x)/(x - 4)

y_right2 = (3*((x+h)-2)) / ((x+h) - 2)
y_left2 =  (3*((x-h)-2)) / ((x-h) - 2)

print("y_right2 = ", round(y_right2,1))
print("y_left2= ", round(y_left2,1))

if round(y_right2) != round(y_left2):
    print("Limit does not exist at x = ", x)
```

### Limits - simple example to compare

```python
x = 2
h = 0.0001

# for y = 3*x

y_right = 3*(x+h)
y_left = 3*(x-h)

print("y_right = ", y_right)
print("y_left = ", y_left)

if round(y_right) != round(y_left):
    print("Limit does not exist at x = ", x)
else:
    print("Limit at x = ", x, " is ", round(y_right))
```

### Limits with Sympy<br>
Example function: y = $\frac{3x^2}{x^2 - 4}$

```python
# for equation y = (3*x**2)/(x**2 - 4)
for x_value in [10, 100, 1000, 10000]:
    limit = (3*x_value**2)/(x_value**2 - 4)
    print("x = ", x_value, " Limit = ", limit)
```

```python
import sympy
from sympy import symbols

x = symbols('x')

# Put the equation here
eq = (3*x**2)/(x**2 - 4)

# critical values
x_1 = 2
x_2 = -2

# limit(expression, variable, value)

# first critical point
right = sympy.limit(eq, x, x_1, dir='+')
left = sympy.limit(eq, x, x_1, dir='-')
print("at x =", x_1)
print("right = ", right)
print("left = ", left)

# second critical point
right = sympy.limit(eq, x, x_2, dir='+')
left = sympy.limit(eq, x, x_2, dir='-')
print("at x =", x_2)
print("right = ", right)
print("left = ", left)

# infinity
right = sympy.limit(eq, x, sympy.oo)
left = sympy.limit(eq, x, -sympy.oo)
print("as x approaches infinity: ", right)
print("as x approaches negative infinity: ", left)
```

### Limits and Slope

```python
import matplotlib.pyplot as plt
import numpy as np

# for y = x**2

x_value = 1
y_value = x_value**2

# change h value to get better answer
h = 0.3
y_value_2 = (x_value + h)**2

zoom = 10
xmin = x_value - zoom
xmax = x_value + zoom
ymin = y_value - zoom
ymax = y_value + zoom

x = np.linspace(xmin, xmax, 100)
y = x**2

slope = (y_value_2 - y_value)/h
print("slope = ", slope)

plt.axis([xmin,xmax,ymin,ymax]) # window size
plt.plot(x, y)
plt.plot([x_value],[y_value],'ro')
plt.plot([x_value + h],[y_value_2],'ro')
plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of y=x^2")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import ipywidgets as widgets
# for y = x**2
def limitslope(h):

        x_value = 1
        y_value = x_value**2

        # change h value to get better answer
        h = h
        y_value_2 = (x_value + h)**2
        
        zoom = 10
        xmin = x_value - zoom
        xmax = x_value + zoom
        ymin = y_value - zoom
        ymax = y_value + zoom
        
        x = np.linspace(xmin, xmax, 100)
        y = x**2
        
        slope = (y_value_2 - y_value)/h
        print("slope = ", slope)
        
        plt.axis([xmin,xmax,ymin,ymax]) # window size
        plt.plot(x, y)
        plt.plot([x_value],[y_value],'ro')
        plt.plot([x_value + h],[y_value_2],'bo')
        plt.axhline(y=0, color='k')
        plt.axvline(x=0, color='k')
        plt.xlabel("x")
        plt.ylabel("y")
        plt.title("Graph of y=x^2")
        plt.show()
interactivesl=widgets.interact(limitslope,h=(0.001,1))
interactivesl
```

$$y=x^2 ---- f(X)=x^2$$

$$point1( x , f(x) ) , point2( x+h , f(x+h)) $$ ~
$$point1( x , y )) , point2( x+h , y) $$ ~
$$point1( x , x^2 )) , point2( x+h, x^2) $$ ~

$$slope=\frac{( x+h )^2- x^2 }{ x+h-x} $$ ~
$$slope=\frac{x^2 + 2xh + h^2 - x^2}{h}$$ ~
$$slope=\frac{2xh + h^2}{h}$$
$$slope=\frac{h(2x + h)}{h}$$
$$slope=2x+h$$
$$\lim_{h \to 0} 2x+h$$

##### proof of logatrithm laws

prove: $$log_{b}^{MN}=log_{b}^{M}+log_{b}^{N}$$
Let  : $$x=log_{b}^{M} $$
$$y=log_{b}^{N} $$
Then : $$M=b^{x}$$
 $$N=b^{y}$$
Thus : $$MN=b^{x}b^{y} $$
 $$MN=b^{x+y}$$
Take log base b on sides:
 $$MN=b^{x+y}$$
 $$log_{b}^{MN}=log_{b}^{b^{x+y}}$$
 $$ log_{b}^{MN}=x+y $$
because $$x=log_{b}^{M} |  | y=log_{b}^{N}$$
$$log_{b}^{MN}=log_{b}^{M}+log_{b}^{N}$$

prove           : $$ log_{b}^{MK}$$
let             : $$ x=log_{b}^{M}$$
then            : $$ M^{K}=b^{x^{k}}$$
log             : $$ log_{b}^{M^{K}}=log_{b}^{b^{K^{X}}}  $$
simplified      : $$ log_{b}^{M^{K}}=K^{X} $$
                : $$ log_{b}^{M^{K}}=KX $$    
substitution    : $$ x=log_{b}^{M}$$
                : $$ log_{b}^{MK}=Klog_{b}^{M} $$

```python
fig,ax=plt.subplots(figsize=(15,15))
plt.plot(np.cos(np.linspace(0,2*np.pi,100)),
         np.sin(np.linspace(0,2*np.pi,100)),alpha=0.5,color='indigo',ls='-.'
        )

plt.axis('square')
plt.plot([0,0],[-1,1],'k--')
plt.plot([-1,1],[0,0],'k--')

plt.plot(np.cos(0.7),
         np.sin(0.7),
         'ro',
         ms=20)

plt.plot([np.cos(0.7),np.cos(0.7)],[0,np.sin(0.7)],color='blue',
         linewidth=5,label='sin(x)/opp')
plt.plot([0,np.cos(0.7)],[0,np.sin(0.7)],
         linewidth=5,label='radius/hyp')
plt.plot([0,np.cos(0.7)],[0,0],color='red',
         linewidth=5,label='cos(x)/adj')
plt.plot([0,0],[0,np.sin(0.7)],color='black',
         linewidth=5,label='sin(x)')
plt.plot([0,np.cos(0.7)],[2*np.sin(0.7),np.sin(0.7)],
         linewidth=5,label='vector(x)')
plt.plot([0,np.cos(0.7)],[np.sin(0.7),np.sin(0.7)],
         ls='--',color='blue',
         linewidth=5,
         label='-sin(x) ')

plt.plot([0,0],[2*np.sin(0.7),np.sin(0.7)],color='red',
         ls='--',
         linewidth=5,label='cos(x)')

ax.set_ylim([-1,1.5])

plt.title('derivatives of sin and cos',
          fontsize=50,
          fontweight='bold',
          color='blue')


plt.annotate('vector-the derivative of position\n is the volecity',
             color='brown',
             fontsize=20,
             xy=(0,np.sin(0.7)*2),
             xytext=(0-0.7,np.sin(0.7)*2+0.1),
             arrowprops=dict(facecolor='red',
                                            edgecolor='purple',
                                            lw=5,
                                            arrowstyle="<-",                                                                    
                                            connectionstyle='arc3,rad=0.001')) 
plt.text(-.95,-0.45,fontsize=20,s='as the arrow is pointing at left direction, then sin is negative')
plt.text(-.95,-0.35,fontsize=20,s='using congruent angle principle to deduce that blue line')
plt.text(-.95,-0.25,fontsize=20,s='and red line -corresponding to sin cos  ')
plt.text(-.95,-0.15,fontsize=20,s='has been flipped 90 degrees the dashed blue line is -sin  ')
plt.text(-.95,-0.05,fontsize=20,s='as the direction is negative and the dashed red line is cos  ')

plt.text(-.95,-0.95,fontsize=20,s='red dot s coordinates are x=cos(x) y=sin(x)')
plt.text(-.95,-0.85,fontsize=20,s='speed = 2 $ \pi $ *radius (1) =2 $ \pi $',color='red')
plt.text(-.95,-0.75,fontsize=20,color='green',s='derivative of sint=sin( t+ pi/2) derivative of cost=cos(yt+pi/2) ')
plt.text(-.95,-0.65,fontsize=20,s='has been flipped 90 degrees the dashed blue line is -sin  ')
plt.legend()
plt.savefig('why derivative of cos equal -sin.pdf')
plt.show()
```

$$ cos(x+sin(x))+a $$

```python
a = sym.symbols('a')

f = sym.cos(x + sym.sin(x)) + a

colors = 'brkm'


for ai in range(0,4):
  if ai==0:
    p = symplot(f.subs(a,ai),show=False,label='a=%s' %ai)
  else:
    p.extend( symplot(f.subs(a,ai),line_color=colors[ai],show=False,label='a=%s' %ai) )

p.title = 'The functions f(x)=cos(x+sin(x))+a'
p.legend = True
p.show()


for ai in range(0,4):
  if ai==0:
    p = symplot(sym.diff(f.subs(a,ai)),show=False,label='a=%s' %ai)
  else:
    p.extend( symplot(sym.diff(f.subs(a,ai)),line_color=colors[ai],show=False,label='a=%s' %ai) )

p.title = "f(x)=cos(x+sin(x))+a 's derivatives"
p.legend = True
plt.savefig('xx.pdf')
p.show()
```

##### ---grahping a hyperbola 

$$y=cosh(x) $$
$$ cos(h)= \frac{e^{x}-e^{-x}}{2}$$

```python
xgph=np.linspace(-3,3,500)
y1=(np.e**xgph)/2
y2=-((np.e**(-xgph))/2)
yhsin=y1+y2
y2l=list(y1)
print(f"The max value in Y list is :{max(y2l)}")
fig,ax=plt.subplots(figsize=(10,8))
ax.hlines(0,-3,3,ls='--',color='cyan')
ax.vlines(0,-10,10,ls='-.',color='indigo')
plt.plot(xgph,y1)
plt.plot(xgph,y2)
plt.plot(xgph,yhsin,color='black')
plt.show()
```

```python
yhcos=y1+y2
y2l=list(y1)
print(f"The max value in Y list is :{max(y2l)}")
fig,ax=plt.subplots(figsize=(10,8))
ax.hlines(0,-3,3,ls='--',color='cyan')
ax.vlines(0,0,10,ls='-.',color='indigo')
plt.plot(xgph,y1)
plt.plot(xgph,y2)
plt.plot(xgph,yhcos,color='black')
plt.show()
```

```python
yhtan=yhsin/yhcos
fig,ax=plt.subplots(figsize=(10,8))
ax.hlines(0,-3,3,ls='--',color='cyan',label="horizontal center line")
ax.vlines(0,-1.5,1.5,ls='-.',color='indigo',label="vertical center line")
ax.hlines(1,-3,3,ls='--',color='crimson',label="top limit")
ax.hlines(-1,-3,3,ls='--',color='maroon',label="bottom limit")
plt.plot(xgph,yhtan,color='black',label="y = tanh(x)")
plt.legend()
plt.show()
```

when taking the derivative of y , it becomes dy with respect to dx
dy means derivative of y -which is the change in y caused by the

### concept

upward projectile  (parabola) 
---equation -  
$$ position(location..of..an..object--time.. x.. y.. height.. xy.. coordinates.. $$
$$position- y=-4.9x^{2}-Vx+h$$
y=height x=time/sec V=initial velocity h=initial height 
$$ velocity.. is.. the.. rate.. of.. change..in.. position.. -derivative.. of.. position$$
$$velocity-y'=-9.8x+V $$
--------set y'=0 to figure out the maximum position (height)
$$0=-9.8x+V $$
-9.8 gravity works against the velocity 
$$ acceleration.. is.. the.. rate.. of..change.. in.. velocity.. whether.. is.. constant.. or.. inc.. or.. decre$$
$$acceleration-y''=-9.8 $$

```python
sympy.plot(position,markers=[{"args":[1.68367346938776,17.7903061224490,"ro"]}])
```

### velocity(derivative of height/position) 

```python
velocity=sympy.diff(position)
print(f"Volecity is : {velocity}")
```

### velocity plot

```python
sympy.plot(velocity,
           markers=[{"args":[1.68367346938776,0,"ro"]},{"args":[0,16.5,"bo"]}])
```

### x value to find max height

```python
x_max_height=sympy.solve(velocity)
max_height=position.subs(t,x_max_height[0])
print(f"max height at {round(x_max_height[0],2)} is seconds ={round(max_height,2)}")
print(f"max height at {int(x_max_height[0])} is seconds ={int(max_height)}")
```

```python
acceleration=sympy.diff(velocity)
print(f"acceleration is {round(acceleration,2)}")
```

### plot-all

```python
sympy.plot(velocity,position,acceleration,
           ylim=[-70,70],
           title="yellow:height,--blue:volecity,--green:accelation\nposition=-4.9*t**2+v*t+h",
           markers=[{"args":[1.68367346938776,0,"ro"]},
                    {"args":[0,16.5,"bo"]},
                    {"args":[1.68367346938776,17.7903061224490,"ro"]}])
plt.savefig('d2.pdf',dpi=5000)
plt.show()
```

## partial derivatives

#### def

partial derivatives method is used when you have more than 1 variable
$$f(x,y)=2xy^2 $$
partial derivative of x
$$ \frac {\partial f }{\partial x} =f_x=2y^2$$
partial derivative of y
$$ \frac {\partial f }{\partial y} =f_y=4xy$$

#### sympy.diff(f,variable)

```python
from sympy.abc import x,y

f=x**2+x*y**2

display(Math('\\frac{ \\partial f}{ \\partial x} =%s'%sympy.latex(sympy.diff(f,x))))
display(Math('\\frac{ \\partial f}{ \\partial y} =%s'%sympy.latex(sympy.diff(f,y))))
```

## derivative vs differential

derivative :
~~ slope of the equation 
$$ y=x^{2} $$
$$ y'/\frac{dy}{dx}=2x$$
differential : 
~~ split it up
$$ dy=2x dx $$
linear approximation - to find out what's the infinitesmal change in x 

#### reason for differential

```python
# reason 
#  to find the linear approximation 
```

#### example

$$ y=x^{2} $$
$$ y'=2x $$
$$ \frac{dy}{dx}=2x $$
$$ dy=2x dx $$

#### steps 

##### 1 derivative of equation

##### solve dy

```python
import sympy
from sympy import symbols 
x,y=symbols('x y')
expression=x**2
der=sympy.diff(expression)
x_val=2
dx=0.05
dy=expression.subs(x,x_val+dx)-expression.subs(x,x_val)
slope=der.subs(x,x_val)
print(f"the infinitesmal change in y is : {round(dy,5)}")
print(f"slope is : {slope}")
print(f"dy/dx : {round(dy/dx,5)}")
print(f"dy=: {round(slope*dx,5)}")
```

#### plot

```python
expression=x**2
sympy.plot(expression,der,
           ylim=[0,10],
           xlim=[-8,8],
           markers=[{"args":[dx,dy,"ro"]}])
```

```python
# Compute critical points using sympy

x = sym.symbols('x')
fx = x**2 * sym.exp(-x**2)

# derivative in sympy, solve
dfx = sym.diff(fx,x)
critpoints = sym.solve(dfx)
print('The critical points are: ' + str(critpoints))

# plot the function derivative and its critical points
y = sym.lambdify(x,dfx)
xx = np.linspace(-3,3,200)

plt.plot(xx,y(xx))
plt.plot([-3,3],[0,0],'k--')
plt.xlim=([-5,5])

for i in critpoints:
    plt.plot(i,0,'ro')

plt.title('Function derivative')
plt.show()
```

## Tangent line's equation

#### concept

$$ y=x^{2} $$
$$ ~y'|\frac{dy}{dx}=2x $$
plug in x value to solve y and slope 
$$x=2 --pluged ..in- - slope=2x=4  $$

$$x=2 --pluged ..in- -  y=2^{2}=4   $$
reconstruct the equation of slope 
$$ y=mx+b $$
$$ 4=4*2+b $$
$$ 4=8+b $$ 
$$ b=-4 $$
final form 
$$ y=4x-4 $$

-------------------------
tangent lines formula 
$$t= f'a(x-x_a) + fa $$
reconstructed 
$$y=mx+b $$
-------f'=m-----------------
$$ f' \text{derivative of a function which is the slope of the function }$$
$$ \text{slope of a line - m} $$
------------------x -- (x-xa) $$ $$
x is shifted by xa (tiny amount) $$ $$
----------------fa is the value of the function --y intercept b 
$$ $$
calculating y intercept fa or b is to set x=0

##### formula

```python
x=symbols('x')
f=x**2
df=sympy.diff(f)

# value at which to comput the tangent line 
xa=0  # point is not fixed - 
fa=f.subs(x,xa)
dfa=df.subs(x,xa)

xx=np.linspace(-5,5,200)
f_fun=sympy.lambdify(x,f)(xx)
df_fun=sympy.lambdify(x,df)(xx)

# contangentline 
tanline=dfa*(xx-xa)+fa
plt.plot(xx,f_fun,label='f(x)')
plt.plot(xx,tanline,label='tangent')
plt.plot(xa,fa,'ro',ms=10)
plt.axis([-10,10,-25,25])
# plt.plot([1,1],[-1.5,1],ls='--')
# plt.plot([-2,1],[1,1],ls='--')
plt.title("'tangentline=df'a*(x-xa)+fa a=0'")
plt.show()
```

```python
x=symbols('x')
f=x**2
df=sympy.diff(f)

# value at which to comput the tangent line 

xx=np.linspace(-2.9,3.9,2000)
f_fun=sympy.lambdify(x,f)(xx)
df_fun=sympy.lambdify(x,df)(xx)

# contangentline 
plt.plot(xx,f_fun,label='f(x)',color='cyan')
for i in np.linspace(-2,3,380):
    tanline=2*i*(xx-i)+i**2
    plt.plot(xx,tanline,label='tangent',linewidth=1,color='cyan')
plt.axis('square')
plt.axis([-2,2,-1,3])
plt.axis('off')
plt.show()
```

```python
# make a function that computes the tangent line, loop through lots of points
#------------------|-f-function #------------------|--------------------|
##-----------------|---xa-point at which tangentline is at #------------|
#------------------|--------|-bnds boundary #a list of 2 elements----- -|
def computetangent(f,  xa,  bnds):
  # define function and derivative values at that point
  df = sym.diff(f)
  fa = f.subs(x,xa)    # using sympy's substitution function is to assume the passed
  df_a = df.subs(x,xa) # parameter argument will be xa xa 

  # evaluate the tangent line
  xx = np.linspace(bnds[0],bnds[1],200)
  return df_a * (xx - xa) + fa
```

```python
x = sym.symbols('x')
f = x**2                       #function 
xx = np.linspace(-2,2,200)     #boundary 
f_fun  = sym.lambdify(x,f)(xx)
                         # boundary 
                        # a
for i in xx:         # function
  yy = computetangent(f,i,xx[[0,-1]])
  plt.plot(xx,yy,color=[abs(i)/3,abs(i)/4,abs(i)/2])

# plot it!
plt.plot(xx,f_fun,'magenta',linewidth=2)

plt.axis('square')
plt.axis([-2,2,-1,3])
plt.axis('off')
plt.show()
```

### Derivative at a point

```python
# first, choose an x value
x_value = 1
step_1 = sympy.Eq(eq,0)
step_2 = step_1.subs(x,x_value)
step_3 = sympy.solve(step_2,y)

# loop because multiple y values are possible
for a in step_3:
    print("x = ", x_value)
    print("y = ", a)
    print("slope = ",derivative.subs(x,x_value).subs(y,a))
    print(" ")
```

### Good output dy/dt and dx/dt<br>

Example: x<sup>2</sup> + y<sup>2</sup> = 25



```python
import sympy
from sympy import symbols, Derivative
x, y, t = sympy.symbols('x y t')

# 0 = expression
expr = x**2 + y**2 - 25

# to see the expression, make this the last line
expr

# to see the derivative
der = expr.diff(x)*Derivative(x,t) + expr.diff(y)*Derivative(y,t)
der
```

substitute for dx/dt and dy/dt

```python
import sympy
from sympy import symbols, Derivative, solve
x, y, t = sympy.symbols('x y t')

# broken down so the substitution works
expr = x**2 + y**2 - 25
expr
dx = expr.diff(x)
dy = expr.diff(y)
dxdt = Derivative(x,t)
dydt = Derivative(y,t)
der = dx*dxdt + dy*dydt
der

# substitute in next code block
```

```python
# this works to sub for dxdt
der.subs(dxdt,3)
```

```python
# plug in all unknowns but one
plugin = der.subs([(dx,6),(dy,8),(dxdt,2)])
plugin
```

```python
# solve for dydt
eq = sympy.Eq(plugin,0)
answer = sympy.solve(eq,dydt)
answer
```

### Volume Example<br>
 V = $\frac{4}{3} \pi$ r<sup>3</sup>

```python
# another example of related rates
import sympy
from sympy import symbols, Derivative
v, r, t = sympy.symbols('v r t')

expr = (4/3)*sympy.pi*r**3
expr
```

```python
# broken down so subtitution works
dr = expr.diff(r)
drdt = Derivative(r,t)
der = dr*drdt
der
```

```python
# plug in r value
dr_plugin = dr.subs(r,4)
plugin = dr_plugin*drdt
plugin
```

```python
# solve for drdt
dvdt = 9
eq = sympy.Eq(plugin,dvdt)
answer = sympy.solve(eq,drdt)
answer
```

### Angle Speed Example:<br>
Camera angle x moving to follow an object<br>
c<sup>2</sup> = a<sup>2</sup> + b<sup>2</sup> * 2ab * cos(x)<br>
Distance c is changing, as the object's speed


```python
import sympy
from sympy import symbols, Derivative
x, c, t = sympy.symbols('x c t')

# a and b are constants
a = 10
b = 10
expr_right = a**2+b**2-2*a*b*sympy.cos(x)
expr_right
```

```python
dx = expr_right.diff(x)
dxdt = Derivative(x,t)
der_right = dx*dxdt
der_right
```

```python
# c
expr_left = c**2
expr_left
```

```python
der_left = expr_left.diff(c)
dcdt = Derivative(c,t)
der_left = der_left*dcdt
der_left
```

```python
# plugin
angle = 0.3
speed = 3
c_value = 2
left = der_left.subs([(dcdt, speed), (c, 2)])
right = dx.subs(x, angle)*dxdt

full_equation = sympy.Eq(left, right)
full_equation
```

```python
answer = sympy.solve(full_equation, dxdt)
answer
```

## mean value theorem

the theorem states that if f is a continuous function on the closed interval [a,b] and differentiable on the open interval (a,b), then there exists a point c in (a,b) such that the tangent at c is parallel to the secant line through the endpoints (a,f(a)) and {\displaystyle (b,f(b))}, that is,

```python
# the theorem states 
#    that if f is a continuous function on the closed interval [a,b] 
#               and differentiable on the open interval (a,b), 

#        then there exists a point c in (a,b) such that
#             the tangent at c is parallel to the secant line 
#                    hrough the endpoints (a,f(a)) and {\displaystyle (b,f(b))}, 
#                         that is,
```

$$f'(c)=\frac{f(b)-f(a)}{b-
a} $$
derivative of point c (xc yc)
$$f'(c) $$
derivative of point b and a 
$$f'(ab)=\frac{f(b)-f(a)}{b-
a} $$
$$ f'(c)=f'(a b) $$

```python
x,y=symbols('x y')
expression=x**3+4*x   

xa=-1
xb=1

fa=expression.subs(x,xa)
fa
```

```python
exderi=expression.diff(x)
```

```python
fb=expression.subs(x,xb)
fb
```

```python
slopemeanv=(fb-fa)/(xb-xa)           # rise two point over run 
slopemeanv
```

```python
sympy.Eq(exderi,slopemeanv)
```

```python
exderi=expression.diff(x)
 # derivative slope must be tangent with ab slope

meanv=solve(sympy.Eq(exderi,slopemeanv),x)
ablxy=[(xa,fa),(xb,fb)]
meanv
# linearequation=
```

```python
sympy.plot(expression,exderi,
           # expression,(x,-1,1),
           ylim=[-100,100],
           xlim=[-10,10],
           markers=[{"args":[meanv[0],0,"ro"]},
                    {"args":[meanv[0],0,"bo"]},
                    {"args":[1,5,"bv"]}])
# sympy.plot(expression,(x,-1,1))
```

### Rolle's Theorem

If a real-valued function f is continuous on a proper closed interval [a, b], differentiable on the open interval (a, b), and f (a) = f (b), then there exists at least one c in the open interval (a, b) such that
f'(c)=0.

```python
# If a real-valued function f 
#        is continuous on a proper closed interval [a, b], 
#     differentiable on the open interval (a, b), 
#     and f (a) = f (b), 
#        then there exists at least one c in the open interval
#          (a, b) such that f'(c)=0.
```

```python
x,y=symbols('x y')
expression=x**2-2*x+1  

xa=-1
xb=3

fa=expression.subs(x,xa)
fb=expression.subs(x,xb)
slopemeanv=(fb-fa)/(xb-xa)           # rise two point over run 
sympy.Eq(exderi,slopemeanv)

exderi=expression.diff(x)
 # derivative slope must be tangent with ab slope

meanv=solve(sympy.Eq(exderi,slopemeanv),x)
ablxy=[(xa,fa),(xb,fb)]
meanv
# linearequation=

sympy.plot(expression,exderi,
           # expression,(x,-1,1),
           ylim=[-100,100],
           xlim=[-10,10],
           markers=[{"args":[meanv[0],0,"ro"]},
                    {"args":[meanv[0],0,"bo"]},
                    {"args":[-1,4,"bv"]},
                    {"args":[3,4,"yv"]}])
# sympy.plot(expression,(x,-1,1))
```

#### method 2find_peakes(exp)

```python
x=symbols('x')
x=np.linspace(-5,5,1001)
fx=x**2*np.exp(-x**2)
dfx=np.diff(fx)/(x[1]-x[0])

# np.diff(fx) only returns df 
# dx is the increment between x[1] and x[0]
# -========however when plotting - np.diff is 1 smaller in numer than  --
# why one less, because np.diff is comparing increment - two ajacent elements derive 1 increment value 
# - therefore, the number of data elements are one less than the one in the original array 
#  the original function , therefore, the original function should be sliced up
#  to the last element to comply with identidal datashape for plotting 

localmax=find_peaks(fx)[0]   # index only  
localmin=find_peaks(-fx)[0]  # index only 

plt.title('find_peaks(function)')
plt.plot(x,fx)
plt.axis([-4,4,0,0.5])

# plot out max and min point
#   ------x is to return the x value which is associated with x axis 
# -------- the find_peaks() method only returns an index 
# ----------therefore, it s necessary to plugin x array to return the x value 
#---------y is the corresponding height at the x location which is returned by 
# --------the indexes using find_peaks 
plt.plot(x[localmax],fx[localmax],'ro',ms=15)
plt.plot(x[localmin],fx[localmin]+0.015,'bo',ms=15)

plt.plot(x[:-1],dfx)
plt.savefig('n.pdf')
plt.axis([-4,4,-0.7,0.7])
plt.savefig('s222d.pdf')
plt.show()
x[localmax]
```

#### ?????critical point exercise

Determine which values of a give the function 
a critical value at x=1 or x=2

$$f(x,a)=x^2e^{-ax^2} $$
$$a {0,0.25,-------2} $$

```python
import sympy as sym
import matplotlib.pyplot as plt
import numpy as np
from scipy.signal import find_peaks
from IPython.display import display,Math

# what values of 'a' give this function a critical point at x=1 or x=2?
a,x = sym.symbols('a,x')

baseexpr = x**2 * sym.exp(-a*x**2)
arange = np.arange(0,2.25,.25)  # a range for evaluation 
xrange = np.linspace(-3,3,100)

# setup plots
fig,ax = plt.subplots(1,2)

for ai in arange:

  fx = baseexpr.subs(a,ai)
  dfx = sym.diff(fx)
  critpnts = sym.solve( dfx )

  # also plot the function in subplot1 and its derivative in subplot2
  ax[0].plot(xrange,sym.lambdify(x,fx)(xrange))
  ax[1].plot(xrange,sym.lambdify(x,dfx)(xrange))


  if 1 in critpnts:
    display(Math('\\Rightarrow %s\\text{ has a critical point at x=1! Woohoo!!}' %sym.latex(fx)))
  elif 2 in critpnts:
    display(Math('\\Rightarrow %s\\text{ has a critical point at x=2! Woohoo!!}' %sym.latex(fx)))
  else:
    display(Math('\\quad %s\\text{ has NO critical point at x=2. :}' %sym.latex(fx)))



# some adjustments to the function plot
ax[0].set_ylim([0,2])
ax[0].set_title('Function')
ax[0].plot([1,1],[0,2],'--',c='gray')
ax[0].plot([2,2],[0,2],'--',c='gray')

# adjustments to the derivative plot
ax[1].set_ylim([-1.5,1.5])
ax[1].plot(xrange[[0,-1]],[0,0],'--',c='gray')
ax[1].plot([1,1],[-1.5,1.5],'--',c='gray')
ax[1].plot([2,2],[-1.5,1.5],'--',c='gray')
# ax[1].set_xlim([.5,2.5])
ax[1].set_title('Its derivative')
fig.set_size_inches(8,4)
plt.savefig('x.pdf')
plt.show()
```

## Area between two curves

#### formula

area under a curve 
$$\int_a^b f(x) dx $$

area under a curve rewrite 
$$\int_a^b [f(x)-0] dx $$

area between two curves 
$$\int_a^b [f(x) - g(x)] dx $$

```python
x = sym.symbols('x')
### expression write out 
symf = x**3
symg = 2*x
# solving the interception points 

eq1=Eq(symf,symg)    
display(Math((sympy.latex(solve(eq1))))) # 
#  assign two expression equal to each other 
# and then solve - the value can be used for fill in x- input data points 

## lambify is to create function that can be called within the fill in - used for defining 
#   the between area y0 and y1 which are corresponding to f and g respectively 
f = sym.lambdify(x,symf)
g = sym.lambdify(x,symg)


# show area using fill_between()
fig, ax = plt.subplots()
xinter = np.linspace(0,np.sqrt(2),100)
ax.fill_between(xinter,f(xinter),g(xinter),color='cyan',alpha=1,label='Area between')
#        syntax x      y0       y1
# fill_ x where on x between y0 and y1
plt.plot(xx,f(xx),label='$f(x)=%s$'%sym.latex(symf))
plt.plot(xx,g(xx),'r',label='$g(x)=%s$'%sym.latex(symg))

plt.legend()
plt.axis([-.25,2,-.1,5])
plt.show()
```

#### latex exercise

### Computing the area between two functions
The area between two functions is given by the formula $A = \int_{a}^{b}f(x) - g(x) dx$

In our example, $f(x)=x^2$ and $g(x)=x$

Therefore, $A = \int_{a}^{b}(x^2 - x) dx$

We will compute the area between the two crossing points, that is, where the two functions are equal. This is given by the two solutions to the equation
$$x^2=x$$

The two solutions are $x=0$ and $x=1$. This gives us the definite integral of
$$A = \int_{0}^{1}(x^2 - x) dx$$

The final answer will be given in the code below.

```python
# find intersection points
fg_intersect = sym.solve(symf-symg)
display(
        Math
         ('f(x)\\text{ and }g(x) \\text{ intersect at } x=%s,%s' %
          (fg_intersect[0],
           fg_intersect[1])))

# compute area and print results!
A = sym.integrate(symf-symg,(x,fg_intersect[0],fg_intersect[1]))
display(Math('\\text{The area between the functions is } A = %s' %A))
```

### workflow

equations 
$$eq1=x^{2} $$
$$eq2=x $$
interception points by equating the two equation 
$$ x^{2}=x $$
$$ point1 =0 ---point2=1 $$
area under line
$$ areaul=\frac{1}{2}1^{2} =\frac{1}{2}$$
area under curve
$$areauc=\int_{0}^{1}\frac{1}{3}x^{3}dx= \frac{1}{3} $$
subtraction 
$$areaul-areauc=abs(\frac{1}{2}-\frac{1}{3})=abs(\frac{3}{6}-\frac{2}{6})=\frac{1}{6}$$

```python
# find the intersection points
x_points = sympy.solve(Eq(eq_1,eq_2)) 
# set equation1 and equation2 equal to each other to solve them 
x_points
```

### Subtract integrals

## Center of Mass

### concept

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

fig, ax = plt.subplots(figsize=(15,10))
polygon = Polygon([(0.2+0.2, 0.3), (0.3+0.2, 0.3), (0.25+0.2, 0.4)], edgecolor='green', facecolor='lightgreen')
ax.add_patch(polygon)

rect=Rectangle((0,0.4),0.1,0.1)
ax.add_patch(rect)

rect=Rectangle((0.9,0.4),0.1,0.1)
ax.add_patch(rect)

ax.set_xlim(0, 1)
ax.set_ylim(0.2, 0.6)
ax.set_aspect('equal')
plt.grid(False)

plt.title('center of mass',fontsize=50)

ax.hlines(0.4,0,1,ls='--',color='red')

plt.show()
```

$$Movment_X=Density\int_a^b\frac{y^{2}}{2}dx $$ 
$$Movment_Y=Density\int_a^bxydx $$ 

center of mass centroid coordinates
$$\bar{X}=\frac{Movement_Y}{Mass} $$ 
$$\bar{Y}=\frac{Movement_x}{Mass} $$ 

```python
import sympy
from sympy import symbols, Eq

x,y,c = symbols('x y c')

# x value range
lower = 0
upper = 9

# for area between eq_1 and x axis
eq_1 = sympy.sqrt(x)

mass = sympy.integrate(eq_1, (x, lower, upper))
eq_2 = (eq_1**2)/2
mx = sympy.integrate(eq_2, (x, lower, upper))
eq_3 = x*eq_1
my = sympy.integrate(eq_3, (x, lower, upper))
x_center = my/mass
y_center = mx/mass
print("x center of mass = ", x_center)
print("y center of mass = ", y_center)
print("point = (", sympy.N(x_center),",", sympy.N(y_center),")")

# plot area with point in the center
sympy.plot(eq_1,(x,lower,upper),markers=[{'args':[x_center,y_center,'ro']}])
```

```python
import sympy
from sympy import symbols, Eq

x,y,c = symbols('x y c')

# x value range
lower = 0
upper = 3.14

# for area between eq_1 and x axis
eq_1 = sympy.sin(x)

mass = sympy.integrate(eq_1, (x, lower, upper))
eq_2 = (eq_1**2)/2
mx = sympy.integrate(eq_2, (x, lower, upper))
eq_3 = x*eq_1
my = sympy.integrate(eq_3, (x, lower, upper))
x_center = my/mass
y_center = mx/mass
print("x center of mass = ", x_center)
print("y center of mass = ", y_center)
print("point = (", sympy.N(x_center),",", sympy.N(y_center),")")

# plot area with point in the center
sympy.plot(eq_1,(x,lower,upper),markers=[{'args':[x_center,y_center,'ro']}])
```

### concept

```python
x=np.linspace(-40,40,500)
y=x**2

fig,ax=plt.subplots(figsize=(15,12))
plt.plot(x,y)

for i in range(0,25,5):
    xp1=5+i
    yp1=(xp1)**2
    
    xp2=10+i
    yp2=(xp2)**2

    plt.text(xp2-2.5,yp1,s='dx',fontsize=15,color='crimson')
    plt.text(xp2,(xp1+2.5)**2,s='dy',fontsize=15,color='cyan')

    plt.scatter(x=xp1,y=yp1,marker='o',color='cyan',s=50)
    plt.scatter(x=xp2,y=yp2,marker='o',color='indigo',s=50)
    
    ax.hlines(yp1,xp1,xp2,ls='--',color='red')
    ax.vlines(xp2,yp1,yp2,ls='-.',color='cyan')
    
    plt.plot([xp1,xp2],[yp1,yp2],ls='--',linewidth=5)



plt.show()


```

### equations

blue dot -a red dot-b
$$a^2+b^2=c^c $$ 
$$ c=\sqrt{a^2+b^2} $$ 
swap a b with dx and dy -change in x and y 
$$ c=\sqrt{(dx)^2+(dy)^2} $$ 
multiply dx^2 to each term within the squareroot radical
$$c=\sqrt{\frac{dx^2}{dx^2}[(dx)^2+(dy)^2]} $$
distribute the dx^2 into the square bracket 
$$c=\sqrt{dx^2(1+(\frac{dy}{dx})^2)} $$
move out the dx^2 from the sqrt radical 
$$c=dx\sqrt{1+(\frac{dy}{dx})^2} $$
intergrate from a to b
$$\int_a^b\sqrt{1+(\frac{dy}{dx})^2} dx $$

```python
import sympy
from sympy import symbols, Eq

x,y,c = symbols('x y c')

# x value range
lower = -3
upper = 3

eq_1 = x**2
der = sympy.diff(eq_1)
eq_2 = sympy.sqrt(der**2 + 1)
arc_length = sympy.integrate(eq_2, (x, lower, upper))
print("arc length = ", arc_length)
print("arc length = ", sympy.N(arc_length))
arc_length
```

## Surface Area

### concept

```python
#  surface area = arc length * circumference 
#  2pi r *arc lenghth- integrated length 
```

```python
import sympy
from sympy import symbols, Eq

x,y,c = symbols('x y c')

# x value range
lower = 0
upper = 3

# boundary equation (y value is the radius)
eq_1 = x**2

# arc length equation
der = sympy.diff(eq_1)
eq_2 = sympy.sqrt(der**2 + 1)

# surface area = 2*pi*radius*arc length
surface_area = sympy.integrate(2*sympy.pi*eq_1*eq_2, (x, lower, upper))

# revolve around x axis
print("surface area = ", sympy.N(surface_area))
print("surface area = ", surface_area)
surface_area
```

### Central Limit Theorem

```python
#1 if sample size is sufficiently large 
              #  then  mean of sample mean distribution = population mean 
                    # ===
                                                         # --------------
#2                variance    
```

```python
# central limit theorem ----(Statistics) statistics

# the fundamental result 
#   that the sum (or mean) of independent identically distributed random variables
#    with finite variance 
#       approaches a normally distributed random variable 
#          as their number increases, 
#            whence in particular if enough samples are repeatedly drawn 
#                from any population,
#                   the sum of the sample values can be thought of, 
#                      approximately, as an outcome 
#                             from a normally distributed random variable
```

```python
ns_imsh=np.arange(50,10000,100)
nbinsim=50

#----------------------------------------------------------#
# empty matrix 100 rows 50 columns # (100, 50
alhist_im=np.zeros((len(ns_imsh),nbinsim))

for i in range(len(ns_imsh)): #100 
    
    #generate numpy array - using sine function 
    sig_1im1=np.sin(np.linspace(0,6*np.pi
                               ,ns_imsh[i]))
    
    #generate normal distribution array 
    sig_1im2=(np.random.rand(ns_imsh[i])*2-1) 
    
    # histogram method to sample from two arrays combined 
    # 
     alhist_im[i,:], x_im1 =np.histogram(sig_1im1+sig_1im2,  
                                        nbinsim,
                                        density=True)

plt.pcolor(x_im1[:-1], ns_imsh , 
           alhist_im, vmax=0.6,cmap='bone')

plt.xlabel('Data Value')
plt.ylabel('Points in the sample')
plt.show()
```

##### normal 

```python

# workflow of creating a normal distribution mannually, 

# first  step - soft code variable by specifying shift stretch and 
#  number of data point values 

#  second step - first use random module to call randn by passing
#    in  number of instances, randomnormal distribution 
#      and then multiply stretch and add shift to the array of randn

# shift   mean location 
# stretch standard deviation scale 
n_dist_shift1=5          #loc   /mean  -shape doesn't change 
n_dist_stretch1=3        #scale /std   -change the countour 
n_dist_n1=1000 

n_dist_pnts1=np.random.randn( n_dist_n1 ) * n_dist_stretch1 + n_dist_shift1
print('Mean is %g, std is %g'%(np.mean(n_dist_pnts1),np.std(n_dist_pnts1)))

fig,ax = plt.subplots(1,2,figsize=(6,4))

ax[0].plot(n_dist_pnts1,'o',alpha=0.4)
ax[1].hist(n_dist_pnts1,bins=70)
ax[1].vlines(5,0,50,ls='--',color='indigo')
ax[0].hlines(5,0,1000,ls='--',color='cyan')
plt.show()
```

##### uniform

```python
n_dist_pnts2=np.random.rand( n_dist_n1 ) * n_dist_stretch1 + 0.5*n_dist_shift1
# rand() produces a mean of 0.5 
# on the contrary randn() yields a mean of 0 
# therefore, the shift value should be half of what's supposed to be 
fig,ax = plt.subplots(1,2,figsize=(6,4))

ax[0].plot(n_dist_pnts2,'o',alpha=0.4)
ax[1].hist(n_dist_pnts2,bins=[3,3.5,4,4.5,5])
print('Mean is %g, range is %g'%(np.mean(n_dist_pnts1),
                                 np.max(n_dist_pnts1)-np.min(n_dist_pnts1)))
ax[1].vlines(4,0,175,ls='--',color='indigo')
ax[0].hlines(4,0,1000,ls='--',color='cyan')
plt.show()
```

##### poisson

```python
poison_lambda=5   

poi_dist_n1=1000 
colors = []
for i in range(poi_dist_n1):
    colors.append('#%06X' % np.random.randint(0, 0xFFFFFF))
    
poi_dist_pnts1=np.random.poisson(poison_lambda, poi_dist_n1 )
print('Mean is %g, variance is %g'%(np.mean(poi_dist_pnts1),np.var(poi_dist_pnts1)))

fig,ax = plt.subplots(1,2,figsize=(6,4))
for i in range(len(poi_dist_pnts1)):
    ax[0].plot(i,poi_dist_pnts1[i],'*',alpha=0.3,color=colors[i])
ax[1].hist(poi_dist_pnts1,bins=np.arange(0,np.max(poi_dist_pnts1)),
           edgecolor='black')
ax[1].vlines(5,0,50,ls='--',color='indigo')
ax[0].hlines(4,0,1000,ls='--',color='cyan')
plt.show()
# the poisson distribution looks like a gaussion distribution 
#  but the mean is shifted towards right mean is equal to variance 
```

##### coin toss/flip plot

```python
faircoin_flipcoin1=0.5
biascoin_flipcoin1=0.6

nTrials_flipcoin1=10

# initialization -by creating a matrix with zeros 
results_flipcoin1=np.zeros((2,nTrials_flipcoin1))
# H&T for loop to update matrix cells values
for i in range(0,nTrials_flipcoin1): 
    results_flipcoin1[0,i]=np.random.rand() > faircoin_flipcoin1  #1row 
    results_flipcoin1[1,i]=np.random.rand() > biascoin_flipcoin1  #2row
    plt.plot(i,results_flipcoin1[0,i],'o',markersize=25)
    plt.plot(i,results_flipcoin1[1,i]-0.2,'*',markersize=25
             ,markeredgewidth=1 ,
             markeredgecolor='black' )
plt.yticks([0,1],['Tails','Heads'])
plt.xlabel('Flip Counts')
plt.ylim=([-.5,1.5])
plt.legend(['Faircoin Toss','Biascoin Toss'],loc='center',fontsize=20)
plt.show()
```

##### die roll probability

```python
n_weighteddie_roll=15

coutcomes_wd_r=np.zeros(6)
print(coutcomes_wd_r)
theoretical_distribution=np.array([2,1,1,1,1,1])

theoretical_distribution=theoretical_distribution/np.sum(theoretical_distribution)

for i in range(n_weighteddie_roll):

    # roll a 7_sided die
    die_roll1=np.random.randint(0,7)

    if die_roll1==6:
        die_roll1=0   # by force any value other than 0 to be 0 
                      # we achieved weighted die that has twice the chance 
                      # than any other side \s value 
    coutcomes_wd_r[die_roll1]   =   coutcomes_wd_r[die_roll1] +1 
   
plt.bar(range(1,7),coutcomes_wd_r/n_weighteddie_roll,
        color='k',label='Empirical proportion',alpha=0.5)
plt.plot(range(1,7),theoretical_distribution,
         'bs-',ms=10,label='Probabilty dirtsibution',alpha=0.2)
plt.legend()
plt.ylabel('probabilty')
plt.xlabel('number of rolls')

for i in range(1,7):
    plt.text( i,  # [i-1] is to gain access to the first element 0
              coutcomes_wd_r[i-1]/n_weighteddie_roll,
              '%s%%' %int(100*coutcomes_wd_r[i-1]/n_weighteddie_roll),
              color='red',fontsize=12,horizontalalignment='center')
plt.savefig('sd.pdf')    
plt.show()
```

#### formula

$$y=e^{\frac{-4 \ln(2) t^2}{h^2}} $$
$$e^{( - )} $$
e's exponent 
$$ \frac{-4 \ln(2)t^2}{h^2} $$ 

time series
$$t^2 $$
rewrite 
$$y=\text{exp}(\frac{-4 \ln(2)t^2}{h^2}) $$
$\text{h means :half max's full width} $
set y equal to 1
$$1=e^{\frac{-4 \ln(2) t^2}{h^2}} $$
apply ln exponent on both sides
$$\ln(1)=ln( \text{exp}(\frac{-4 \ln(2)t^2}{h^2}) $$
ln(1)=0 
$$ ln_e^{(1)}=0  \quad \log_e^{0}=1 $$
symplifed 
$$0=\frac{-4\ln(2)t^2}{h^2} $$
multiply bothside by $h^2$ 
$$0=-4 \ln(2)t^2 $$
take square root to isolate the t 
$$0=2t\sqrt{-\ln(2)} $$

$$\text{exp}(\frac{-4 \cdot \log{(2)} \cdot t^2)}{h^2}) $$
FWHM
$$h\cdot \sqrt{\frac{\log(\quad yfind \quad)}{-4\cdot\log(2)}} $$

```python
gaussian_t1=np.linspace(-3,3,1001) #odd number of data points will have zero in the middle 
gaussian_h1=0.999

fig,ax=plt.subplots()
#gaussian expression 
gaussian_g1=np.exp(-4*np.log(2)*gaussian_t1**2/gaussian_h1**2)
plt.plot(gaussian_t1,gaussian_g1)

plt.xlabel('t')
plt.ylabel('g')
plt.title('Gaussian with FWHM : '+str(gaussian_h1))

plt.plot([-3,3],[0.5,0.5],ls='--',color='black')
plt.plot([0.5,0.5],[0,0.5],ls='--',color='black')
plt.plot([-0.5,-0.5],[0,0.5],ls='--',color='black')

ax.set_yticks(np.arange(0,1,0.1))
ax.set_xticks(np.arange(-3,3,0.5))

###-------------fill between -0.5 and 0.5 with color cyan -----------
gaussian_t2fill=np.linspace(-0.5,0.5,1001)
gaussian_g2fill=np.exp(-4*np.log(2)*gaussian_t2fill**2/gaussian_h1**2)
plt.fill_between(gaussian_t2fill,gaussian_g2fill,color='cyan')

plt.show()
########################### calculating FWHM ###########
yvalfin=.5
tvalsn=gaussian_h1*np.sqrt(np.log(yvalfin)/(-4*np.log(2) ) )

print(np.round(-tvalsn,1),np.round(tvalsn,1))
#   numpy only returns positive squareroot value 
#    therefore, the negative counterpart should be added mannually 
#        Return the non-negative square-root of an array, element-wise.
```

FWHM

```python
gaussian_t1=np.linspace(-3,3,1001) #odd number of data points will have zero in the middle 
gaussian_h1=0.999
#gaussian plot
fig,ax=plt.subplots(figsize=(15,10))
gaussian_g1=np.exp(-4*np.log(2)*gaussian_t1**2/gaussian_h1**2)

plt.plot(gaussian_t1,gaussian_g1)
plt.xlabel('t')
plt.ylabel('g')
plt.title('Gaussian with FWHM : '+str(gaussian_h1))
plt.plot([-0.5,0.5],[0.5,0.5],ls='--',color='black')
plt.plot([0.5,0.5],[0,0.5],ls='--',color='black')
plt.plot([-0.5,-0.5],[0,0.5],ls='--',color='white')

ax.set_yticks(np.arange(0,1,0.1))
ax.set_xticks(np.arange(-3,3,0.5))

########################### fill between ###########################
gaussian_t2fill=np.linspace(-0.5,0.5,1001)
gaussian_g2fill=np.exp(-4*np.log(2)*gaussian_t2fill**2/gaussian_h1**2)
plt.fill_between(gaussian_t2fill,gaussian_g2fill,color='cyan')
gaussian_t3fill=np.linspace(-3,-0.5,1001)
gaussian_g3fill=np.exp(-4*np.log(2)*gaussian_t3fill**2/gaussian_h1**2)
gaussian_t4fill=np.linspace(0.5,3,1001)
gaussian_g4fill=np.exp(-4*np.log(2)*gaussian_t4fill**2/gaussian_h1**2)

plt.fill_between(gaussian_t2fill,gaussian_g2fill,color='yellow')
plt.fill_between(gaussian_t3fill,gaussian_g3fill,color='indigo')
plt.fill_between(gaussian_t4fill,gaussian_g4fill,color='crimson')
########################### fill between ###########################
fig.set_facecolor('cyan')
ax.set_facecolor('black')
plt.grid(alpha=0.2)

########################### FWHM-calculation ###########
yvalfin=.5
tvalsn=gaussian_h1*np.sqrt(np.log(yvalfin)/(-4*np.log(2) ) )

print(np.round(-tvalsn,1),np.round(tvalsn,1))
#   numpy only returns positive squareroot value 
#    therefore, the negative counterpart should be added mannually 
# Return the non-negative square-root of an array, element-wise.
########################## annotation ####################
cl=matplotlib.patches.Circle((-.5,0.5),0.015,color='green')
cr=matplotlib.patches.Circle((.5,0.5),0.015,color='green')
rtg=matplotlib.patches.Rectangle((.6,0.5),2.5,0.5,color='beige')
ax.add_patch(cl)
ax.add_patch(cr)
ax.add_patch(rtg)

plt.annotate('Full width at \nhalf max\nis y at 0.5 \nx is -0.5 and 0.5',
             color='black',
             fontsize=30,
             xy=(0.5,0.5),
             xytext=(1,0.65),
             arrowprops=dict(facecolor='cyan',
                                            edgecolor='cyan',
                                            lw=3,
                                            arrowstyle="<->",                                                                    
                                            connectionstyle='arc3,rad=0.5'))    

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

#### parameters preset update needed ###
h=0.999
t=np.linspace(-3,3,1001)
N=100

#### initializing Matrix ###################
M=np.zeros((N,len(t)))
h=np.zeros(N)

#### for loop to update values into the matrix's cells ####
for i in range(N):
    h[i]=(i+1)/N # scaledown 100 -from y 100 to y 1 h value increases 
    M[i,:]=np.exp(-4*np.log(2)*t**2/h[i]**2) # fillout all the columns in a single row
    # and then proceed to the next row 
    # why it worked - because the rows are created using separate variabel N 
    # and all the column values are created using np.linspace() to create an numpy array object 
    # or using nested for loop outer for row and inner for columns
#### visualization ###############################################
plt.pcolormesh(t,h,M)
plt.savefig('dd.pdf')
plt.show()
```

#### np.random.normal()

```python
import numpy as np
import matplotlib.pyplot as plt
# Parameters
mean = 0 # μ
std_dev = 1 # σ
size = 1000 # number of samples
# Generate random samples
samples = np.random.normal(mean, std_dev, size)
# Plot histogram and PDF
count, bins, _ = plt.hist(samples, bins=30, density=True, alpha=0.6)
pdf = (1 / (std_dev * np.sqrt(2 * np.pi))) * np.exp(- (bins - mean)**2 / (2 * std_dev**2))
plt.plot(bins, pdf, 'r', linewidth=2)
plt.title("Normal Distribution")
plt.show()
```

#### norm.cdf()

```python
from scipy.stats import norm
mean = 78
std_dev = 25
score = 60
# Cumulative probability (P(X < score))
probability = norm.cdf(score, loc=mean, scale=std_dev)
print(f"Probability: {probability:.4f}")
```

### gradient

***
$\mathbf{\text{Gradient Tree Boosting Algorithm}}$<br>
***
1.&emsp;Initialize model with a constant value $$f_{0}(x) = \textrm{arg min}_{\gamma} \sum \limits _{i=1} ^{N} L(y_{i}, \gamma)$$
2.&emsp;For m = 1 to M:<br>
&emsp;&emsp;(a)&emsp;For $i = 1,2,...,N$ compute<br>
    $$r_{im} = - \displaystyle \Bigg[\frac{\partial L(y_{i}, f(x_{i}))}{\partial f(x_{i})}\Bigg]_{f=f_{m−1}}$$
&emsp;&emsp;(b)&emsp;Fit a regression tree to the targets $r_{im}$ giving terminal regions<br>
&emsp;&emsp;&emsp;&emsp;$R_{jm}, j = 1, 2, . . . , J_{m}.$<br><br>
&emsp;&emsp;(c)&emsp;For $j = 1, 2, . . . , J_{m}$ compute<br>
$$\gamma_{jm} = \underset{\gamma}{\textrm{arg min}} \sum \limits _{x_{i} \in R_{jm}} L(y_{i}, f_{m−1}(x_{i}) + \gamma)$$
<br>
&emsp;&emsp;(d)&emsp;Update $f_{m}(x) = f_{m−1}(x) + \sum _{j=1} ^{J_{m}} \gamma_{jm} I(x \in R_{jm})$<br><br>
3. Output $\hat{f}(x) = f_{M}(x)$
***

```python
# $ : All the Math you want to write in the markdown should be 
# inside opening and closing $ symbol 
# in order to be processed as Math.

# limit
# Limit at plus infinity	: $$\lim_{x \to +\infty} f(x)$$ 
# Limit at minus infinity	: $$\lim_{x \to -\infty} f(x)$$	
# Limit at $\alpha$  : $$\lim_{x \to \alpha} f(x)$$

# Max : $$\max_{x \in [a,b]}f(x)$$	
# Min : $$\min_{x \in [\alpha,\beta]}f(x)$$	
# Sup : $$\sup_{x \in \mathbb{R}}f(x)$$	
# Inf : $$\inf_{x > s}f(x)$$

# Please refer https://towardsdatascience.com/write-
#  markdown-latex-in-the-jupyter-notebook-10985edb91fd for example

# To add little spacing in math mode use \,
# To add a new line when in math mode use \\
# To display fraction use \frac{arg 1}{arg 2}
# For power (superscripts text) use ^{}
# For indices (subscripts) use _{}
# For roots use \sqrt[n]{arg}
# The [n]is optional.


# \beta : Creates the symbol beta

# \hat{} : A hat is covered over anything inside 
    # the curly braces of \hat{}. 
    #     E.g. in \hat{Y} hat is 
            # created over Y and in \hat{\beta}_{0},  
            # hat is shown over beta

# _{} : Creates as subscript, 
    # anything inside the curly braces after _. E.g. 
    # \hat{\beta}_{0} will create beta with a hat
    # and give it a subscript of 0.

# ^{} : (Similar to subscript) Creates as superscript, 
    # anything inside the curly braces after ^.

# \sum : Creates the summation symbol

# \limits _{} ^{} : Creates lower and upper limit 
    #  for the \sum using the subscript and 
    # superscript notation.

# *** : Creates horizontal line

# &emsp; : Creates space. (Ref: Space in ‘markdown’ 
  # cell of Jupyter Notebook)

# \gamma : Creates gamma symbol

# \displaystyle : Forces display mode (BONUS 3 above).
  #(Ref: Display style in Math mode)

# \frac{}{} : Creates fraction with two curly braces
  # from numerator and denominator.

# <br> : Creates line breaks

# \Bigg : Helps create parenthesis of big sizes. 
  # (Ref: Brackets and Parentheses)

# \partial : Creates partial derivatives symbol

# \underset() : To write under a text. E.g. 
  # gamma under arg min, instead of a subscript. 
  # In the algorithm you’ll see both types.

# \in : Creates belongs to symbol which is
  # heavily used in set theory.


# <b> Roman Names </b>
# $$

# \sin(-\alpha)=-\sin(\alpha)\\

# \arccos(x)=\arcsin(u)\\

# \log_n(n)=1\\

# \tan(x) = \frac{\sin(x)}{\cos(x)}

# $$

# First order derivative : $$f'(x)$$ 

# K-th order derivative  : $$f^{(k)}(x)$$

# Partial firt order deivative	: $$\frac{\partial f}{\partial x}$$

# Partial k-th order derivative	: $$\frac{\partial^{k} f}{\partial x^k}$$

# Sum from 0 to +inf:

# $$\sum_{j=0}^{+\infty} A_{j}$$

# Double sum:

# $$\sum^k_{i=1}\sum^{l+1}_{j=1}\,A_i A_j$$	

# Taylor expansion of $e^x$:

#  $$ e^x  = \sum_{k=0}^{n}\, \frac{x^k}{k!} + o(x^n) $$ 

# $$
# \frac{arg 1}{arg 2} \\

# x^2\\

# e^{i\pi}\\

# A_i\\

# B_{ij}\\

# \sqrt[n]{arg}	

# $$

# Greek Letters <br>
# Given : $\pi = 3.14$ , $\alpha = \frac{3\pi}{4}\, rad$
# $$
# \omega = 2\pi f \\
# f = \frac{c}{\lambda}\\
# \lambda_0=\theta^2+\delta\\
# \Delta\lambda = \frac{1}{\lambda^2}
# $$
```

```python
fig,axs=plt.subplots(2,2,figsize=(10,5))
# convas fig axs is the miniture xy axies based on which plots are generated 

fig.set_facecolor('cyan')
axs[0,0].set_facecolor('green')
axs[0,1].set_facecolor('lightblue')
axs[1,0].set_facecolor('indigo')
axs[1,1].set_facecolor('beige')

x=np.linspace(-7,7,100)
ysin=np.sin(x)
ycos=np.cos(x)
ytan=np.tan(x)
ycub=x**3

axs[0,0].plot(x,ysin,color='black')
axs[0,1].plot(x,ycos,color='red')
axs[1,0].plot(x,ytan,color='white')
axs[1,1].plot(x,ycub,color='cyan')

plt.show()
```
